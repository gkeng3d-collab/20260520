```python
# om pickle import TRUE
date_now = datetime.now()
date_formated = date_now.strftime("%Y-%m-%d")
date_csv = date_now.strftime("%Y%m%d")
date_lastyear = datetime.now() - relativedelta(years=1)
st_date_formated = date_lastyear.strftime("%Y-%m-%d")

csvfile_intday = 'Result' + date_csv +'interday' + '.csv'
csvfile = 'Stocklist_' + date_csv + '.csv'
csvfile25_3 = 'Break25_OverVolMAx3' + date_csv + '.csv'



# add table 1st time
# def addtable1info(Stock_df):
#     try:
#         lv_syml_info = yf.Ticker(A)
#         info = lv_syml_info.info
#         if not info:
#             print(f"Ticker {A} ไม่สามารถดึงข้อมูลได้")
#             return
#         marketCap = info.get('marketCap', None)
#         Stock_df['Mcap'] = marketCap
#         pe = info.get('trailingPE', None)
#         Stock_df['PE'] = pe
#         payoutRatio = info.get('payoutRatio', None)
#         Stock_df['payoutRatio'] = payoutRatio
#         dividendYield = info.get('dividendYield', None)
#         Stock_df['dividendYield'] = dividendYield
#         floatShares = info.get('floatShares', None)
#         Stock_df['floatShares'] = floatShares
#         sharesOutstanding = info.get('sharesOutstanding', None)
#         Stock_df['sharesOutstanding'] = sharesOutstanding
#         if floatShares and sharesOutstanding:
#             Stock_df['freefloatShares%'] = (floatShares / sharesOutstanding) * 100
#         else:
#             Stock_df['freefloatShares%'] = None
#         enterpriseValue = info.get('enterpriseValue', None)
#         Stock_df['enterpriseValue'] = enterpriseValue
#     except Exception as e:
#         print(f"Error in addtable1info: {e}")

def addtable1info(Stock_df):
  lv_syml_info = yf.Ticker(A)
  marketCap = lv_syml_info.info.get('marketCap', None)
  Stock_df['Mcap'] = marketCap
  pe = lv_syml_info.info.get('trailingPE', None)
  Stock_df['PE'] = pe

  payoutRatio = lv_syml_info.info.get('payoutRatio', None)
  Stock_df['payoutRatio'] = payoutRatio

  dividendYield = lv_syml_info.info.get('dividendYield', None)
  Stock_df['dividendYield'] = dividendYield

  floatShares = lv_syml_info.info.get('floatShares', None)
  Stock_df['floatShares'] = floatShares
  floatShares = lv_syml_info.info.get('sharesOutstanding', None)
  Stock_df['sharesOutstanding'] = floatShares
  Stock_df['freefloatShares%'] = (Stock_df['floatShares']/Stock_df['sharesOutstanding'])*100

  enterpriseValue = lv_syml_info.info.get('enterpriseValue', None)
  Stock_df['enterpriseValue'] = enterpriseValue

  # try:
  #     Stock_df['PE'] = lv_syml_info.info['trailingPE']
  # except NameError:
  #     Stock_df['PE'] = -999999

def addtable1(Stock_df):
  Stock_df['Name'] = lv_syml
  Stock_df['datetime_'] = Stock_df.index.date
  Stock_df['open-1'] = Stock_df['open'].shift(1)
  Stock_df['high-1'] = Stock_df['high'].shift(1)
  Stock_df['low-1'] = Stock_df['low'].shift(1)
  Stock_df['close-1'] = Stock_df['close'].shift(1)
  Stock_df['close-2'] = Stock_df['close'].shift(2)
  Stock_df['close-5'] = Stock_df['close'].shift(5)
  Stock_df['close-6'] = Stock_df['close'].shift(6)
  Stock_df['close-10'] = Stock_df['close'].shift(10)
  Stock_df['close-11'] = Stock_df['close'].shift(11)
  Stock_df['close-20'] = Stock_df['close'].shift(20)
  Stock_df['close-60'] = Stock_df['close'].shift(60)
  Stock_df['close-75'] = Stock_df['close'].shift(75)
  Stock_df['close-76'] = Stock_df['close'].shift(76)
  Stock_df['close-240'] = Stock_df['close'].shift(240)
  Stock_df['Values'] = Stock_df['volume'] * Stock_df['close']
  Stock_df['retune_chg'] =  100 * ((Stock_df['close']-Stock_df['close-1'])/Stock_df['close-1'])
  Stock_df['retune_log'] =  100 * (np.log(Stock_df['close']) - np.log(Stock_df['close'].shift(1)))
  Stock_df['retune_chg-1'] = Stock_df['retune_chg'].shift(1)
  Stock_df['retune_chg-2'] = Stock_df['retune_chg'].shift(2)
  Stock_df['retune_chg-5'] = Stock_df['retune_chg'].shift(5)
  Stock_df['retune_chg-10'] = Stock_df['retune_chg'].shift(10)
  Stock_df['retune_chg-12'] = Stock_df['retune_chg'].shift(12)
  Stock_df['retune_chg-20'] = Stock_df['retune_chg'].shift(20)
  Stock_df['retune_chg-26'] = Stock_df['retune_chg'].shift(26)
  Stock_df['retune_chg-40'] = Stock_df['retune_chg'].shift(40)
  Stock_df['retune_chg-60'] = Stock_df['retune_chg'].shift(60)
  Stock_df['retune_chg-75'] = Stock_df['retune_chg'].shift(75)
  Stock_df['OpenHigherlastclose'] = Stock_df['open'] > Stock_df['close-1']

  Stock_df['Uper'] = Stock_df['low'] > Stock_df['close-1']
  Stock_df['Open Gap'] = Stock_df['low'] > Stock_df['high-1']
  Stock_df['Jump'] = (Stock_df['OpenHigherlastclose'] == True) & (Stock_df['retune_chg'] > 1.5)
  Stock_df['Jump Gap'] = (Stock_df['OpenHigherlastclose'] == True) & (Stock_df['Open Gap'] == True) &(Stock_df['retune_chg'] > 1.5)
  Stock_df['Jump low'] = Stock_df['high'] < Stock_df['low-1']
  Stock_df['Jump up'] = Stock_df['low'] > Stock_df['high-1']
  Stock_df['Jump low-1'] = Stock_df['Jump low'].shift(1)
  Stock_df['Jump up-1'] = Stock_df['Jump up'].shift(1)
  Stock_df['Jump low-2'] = Stock_df['Jump low'].shift(2)
  Stock_df['Jump up-2'] = Stock_df['Jump up'].shift(2)


  Stock_df['close_st'] = Stock_df['close'].iloc[0]
  # Stock_df['close_st_manual'] = Stock_df['close'].iloc[-13]
  # Stock_df['close_st_manual'] = Stock_df.loc[datestartst_manual, 'close']
  if datestartst_manual in Stock_df.index:
    Stock_df['close_st_manual'] = Stock_df.loc[datestartst_manual, 'close']
  else:
    Stock_df['close_st_manual'] = 0

  Stock_df['return_chg_st'] =  100 * ((Stock_df['close']-Stock_df['close_st'])/Stock_df['close_st'])
  Stock_df['return_log_st'] =  100 * (np.log(Stock_df['close']) - np.log(Stock_df['close_st']))
  Stock_df['return_chg_manual'] = np.where(Stock_df['close_st_manual'] == 0, 1,100 * ((Stock_df['close']-Stock_df['close_st_manual'])/Stock_df['close_st_manual']))
  Stock_df['return_chg_week'] =  100 * ((Stock_df['close']-Stock_df['close-5'])/Stock_df['close-5'])
  Stock_df['return_chg_month'] =  100 * ((Stock_df['close']-Stock_df['close-20'])/Stock_df['close-20'])
  Stock_df['return_chg_q'] =  100 * ((Stock_df['close']-Stock_df['close-60'])/Stock_df['close-60'])
  Stock_df['return_chg_y'] =  100 * ((Stock_df['close']-Stock_df['close-240'])/Stock_df['close-240'])
    #หาytd
  Stock_df['return_chg_ytd'] = 100 * ((Stock_df['close'] - Stock_df['close'].iloc[-YTDx]) / Stock_df['close'].iloc[-YTDx])


def addtablecandlestick(Stock_df):
  Stock_df['candlerange'] = Stock_df['high'] - Stock_df['low']
  Stock_df['candlerangebody'] = Stock_df['close']  - Stock_df['open']
  Stock_df['candletaiLowGreen'] =  Stock_df['open'] - Stock_df['low']
  Stock_df['candletailLowRed'] = Stock_df['close'] - Stock_df['low']
  Stock_df['candletaiupRed'] =  Stock_df['high'] - Stock_df['open']
  Stock_df['candletaiupGreen'] =  Stock_df['high'] - Stock_df['close']



  Stock_df['%candleClose'] = ((Stock_df['close']-Stock_df['low'])/Stock_df['candlerange'])*100
  Stock_df['%candleClose-1'] = Stock_df['%candleClose'].shift(1)
  Stock_df['%candleClose-2'] = Stock_df['%candleClose'].shift(2)
  Stock_df['%candleClose-5'] = Stock_df['%candleClose'].shift(5)
  Stock_df['%candleClose-10'] = Stock_df['%candleClose'].shift(10)

  Stock_df['%candleBody'] = ((Stock_df['close']-Stock_df['open'])/Stock_df['candlerange'])*100
  Stock_df['%candleBody-1'] = Stock_df['%candleBody'].shift(1)
  Stock_df['%candleBody-2'] = Stock_df['%candleBody'].shift(2)
  Stock_df['%candleBody-5'] = Stock_df['%candleBody'].shift(5)
  Stock_df['%candleBody-10'] = Stock_df['%candleBody'].shift(10)

  Stock_df['%candleOpen'] = ((Stock_df['open']-Stock_df['low'])/Stock_df['candlerange'])*100
  Stock_df['%candleOpen-1'] = Stock_df['%candleOpen'].shift(1)
  Stock_df['%candleOpen-2'] = Stock_df['%candleOpen'].shift(2)
  Stock_df['%candleOpen-5'] = Stock_df['%candleOpen'].shift(5)
  Stock_df['%candleOpen-10'] = Stock_df['%candleOpen'].shift(10)


  Stock_df['candleOHLC'] = (Stock_df['close']==Stock_df['low'])&(Stock_df['close']==Stock_df['open'])&(Stock_df['close']==Stock_df['high'])
  Stock_df['candleHC'] = (Stock_df['close']!=Stock_df['open'])&(Stock_df['close']==Stock_df['high'])
  Stock_df['candleLC'] = (Stock_df['close']==Stock_df['low'])&(Stock_df['close']!=Stock_df['open'])
  Stock_df['candleOHC'] = (Stock_df['close']!=Stock_df['low'])&(Stock_df['close']==Stock_df['open'])&(Stock_df['close']==Stock_df['high'])
  Stock_df['candleOLC'] = (Stock_df['close']==Stock_df['low'])&(Stock_df['close']==Stock_df['open'])&(Stock_df['close']!=Stock_df['high'])

  Stock_df['candleDoji'] = (Stock_df['close']==Stock_df['open'])
  Stock_df['candleDoji-1'] = Stock_df['candleDoji'].shift(1)
  Stock_df['candleDoji-2'] = Stock_df['candleDoji'].shift(2)
  Stock_df['candleDoji-5'] = Stock_df['candleDoji'].shift(5)
  Stock_df['candleDoji-10'] = Stock_df['candleDoji'].shift(10)
  Stock_df['candleDoji_Drajonfly'] = ((Stock_df['close']==Stock_df['open'])&(Stock_df['close']==Stock_df['high']))
  Stock_df['candleDoji_Gravstone'] = ((Stock_df['close']==Stock_df['open'])&(Stock_df['close']==Stock_df['low']))
  Stock_df['candleDoji_4price'] = (Stock_df['close']==Stock_df['low'])&(Stock_df['close']==Stock_df['open'])&(Stock_df['close']==Stock_df['high'])

  Stock_df['candle_Hammer_G'] = (Stock_df['close']>Stock_df['open'])&(Stock_df['%candleClose']>60)&(Stock_df['%candleOpen']>60)
  Stock_df['candle_Hammer_G-1'] = Stock_df['candle_Hammer_G'].shift(1)
  Stock_df['candle_Hammer_G-2'] = Stock_df['candle_Hammer_G'].shift(2)
  Stock_df['candle_Hammer_G-5'] = Stock_df['candle_Hammer_G'].shift(5)
  Stock_df['candle_Hammer_R'] = (Stock_df['open']>Stock_df['close'])&(Stock_df['%candleOpen']>60)&(Stock_df['%candleClose']>60)
  Stock_df['candle_Hammer_R-1'] = Stock_df['candle_Hammer_R'].shift(1)
  Stock_df['candle_Hammer_R-2'] = Stock_df['candle_Hammer_R'].shift(2)
  Stock_df['candle_Hammer_R-5'] = Stock_df['candle_Hammer_R'].shift(5)
  Stock_df['candle_ReHammer_G'] = (Stock_df['close']>Stock_df['open'])&(Stock_df['%candleClose']<40)&(Stock_df['%candleOpen']<40)
  Stock_df['candle_ReHammer_G-1'] = Stock_df['candle_ReHammer_G'].shift(1)
  Stock_df['candle_ReHammer_G-2'] = Stock_df['candle_ReHammer_G'].shift(2)
  Stock_df['candle_ReHammer_G-5'] = Stock_df['candle_ReHammer_G'].shift(5)
  Stock_df['candle_ReHammer_R'] = (Stock_df['open']>Stock_df['close'])&(Stock_df['%candleOpen']<40)&(Stock_df['%candleClose']<40)
  Stock_df['candle_ReHammer_R-1'] = Stock_df['candle_ReHammer_R'].shift(1)
  Stock_df['candle_ReHammer_R-2'] = Stock_df['candle_ReHammer_R'].shift(2)
  Stock_df['candle_Bullish_engulfing'] = ((Stock_df['%candleOpen-1'] > Stock_df['%candleClose-1'])&(Stock_df['open']<Stock_df['close-1'])&(Stock_df['close']>Stock_df['open-1']))
  Stock_df['candle_Bullish_engulfing-1'] = Stock_df['candle_Bullish_engulfing'].shift(1)
  Stock_df['candle_Bullish_engulfing-2'] = Stock_df['candle_Bullish_engulfing'].shift(2)
  Stock_df['candle_Bearish_engulfing'] = ((Stock_df['%candleOpen-1'] < Stock_df['%candleClose-1'])&(Stock_df['open']>Stock_df['close-1'])&(Stock_df['close']<Stock_df['open-1']))
  Stock_df['candle_Bearish_engulfing-1'] = Stock_df['candle_Bearish_engulfing'].shift(1)
  Stock_df['candle_Bearish_engulfing-2'] = Stock_df['candle_Bearish_engulfing'].shift(2)
  Stock_df['candle_Bullish_Harami'] = ((Stock_df['%candleOpen-1'] < Stock_df['%candleClose-1'])&(Stock_df['high']<Stock_df['high-1'])&(Stock_df['low']>Stock_df['low']))
  Stock_df['candle_Bullish_Harami-1'] = Stock_df['candle_Bullish_Harami'].shift(1)
  Stock_df['candle_Bullish_Harami-2'] = Stock_df['candle_Bullish_Harami'].shift(2)
  Stock_df['candle_Bearish_Harami'] = ((Stock_df['%candleOpen-1'] > Stock_df['%candleClose-1'])&(Stock_df['open']>Stock_df['close-1'])&(Stock_df['low']>Stock_df['low-1']))
  Stock_df['candle_Bearish_Harami-1'] = Stock_df['candle_Bearish_Harami'].shift(1)
  Stock_df['candle_Bearish_Harami-2'] = Stock_df['candle_Bearish_Harami'].shift(2)






  # if Stock_df['close']>Stock_df['open']:
  #   Stock_df['candleRYG'] = 'G'
  #   Stock_df['%candle_tailLow'] = ((Stock_df['open'] - Stock_df['low'] )/Stock_df['candlerange'])*100
  #   Stock_df['%candle_tailup'] = ((Stock_df['high'] - Stock_df['close'])/Stock_df['candlerange'])*100
  #   Stock_df['%candlerangebody'] = ((Stock_df['close']  - Stock_df['open'])/Stock_df['candlerange'])*100

  # elif Stock_df['close']<Stock_df['open']:
  #   Stock_df['candleRYG'] = 'R'
  #   Stock_df['%candle_tailLow'] = ((Stock_df['close'] - Stock_df['low'] )/Stock_df['candlerange'])*100
  #   Stock_df['%candle_tailup'] = ((Stock_df['high'] - Stock_df['open'])/Stock_df['candlerange'])*100
  #   Stock_df['%candlerangebody'] = ((Stock_df['open'] - Stock_df['close'])/Stock_df['candlerange'])*100
  # else:
  #   Stock_df['candleRYG'] = 'Y'
  #   Stock_df['%candle_tailLow'] = ((Stock_df['close'] - Stock_df['low'] )/Stock_df['candlerange'])*100
  #   Stock_df['%candle_tailup'] = ((Stock_df['high'] - Stock_df['close'])/Stock_df['candlerange'])*100
  #   Stock_df['%candlerangebody'] = '0'


def addtablehikenashi(Stock_df):

  # คำนวณ Heikin Ashi
  Stock_df['HA_Close'] = (Stock_df['open'] + Stock_df['high'] + Stock_df['low'] + Stock_df['close']) / 4

  # สร้างคอลัมน์ HA_Open (ใช้ค่าเริ่มต้น)
  for i in range(Stock_df.shape[0]):
    if i > 0:
      Stock_df.loc[Stock_df.index[i],'HA_Open'] = (Stock_df['HA_Open'][i-1] + Stock_df['HA_Close'][i-1])/2
    elif i == 0:
      Stock_df.loc[Stock_df.index[0],'HA_Open'] = (Stock_df['open'][0] + Stock_df['close'][0])/2

  # คำนวณคอลัมน์ HA_High และ HA_Low
  Stock_df['HA_High'] = Stock_df[['high', 'HA_Open', 'HA_Close']].max(axis=1)
  Stock_df['HA_Low'] = Stock_df[['low', 'HA_Open', 'HA_Close']].min(axis=1)

  Stock_df['HA_Close-1'] = Stock_df['HA_Close'].shift(1)
  Stock_df['HA_Open-1'] = Stock_df['HA_Open'].shift(1)
  Stock_df['HA_High-1'] = Stock_df['HA_High'].shift(1)
  Stock_df['HA_Low-1'] = Stock_df['HA_Low'].shift(1)
  Stock_df['HA_Close_EMA10'] = Stock_df['HA_Close'].ewm(span=3).mean()
  Stock_df['HA_Close_EMA10-1'] = Stock_df['HA_Close_EMA10'].shift(1)
  Stock_df['HA_Close-2'] = Stock_df['HA_Close'].shift(2)
  Stock_df['HA_Open-2'] = Stock_df['HA_Open'].shift(2)
  Stock_df['HA_High-2'] = Stock_df['HA_High'].shift(2)
  Stock_df['HA_Low-2'] = Stock_df['HA_Low'].shift(2)
  Stock_df['HA_Close_EMA10-1'] = Stock_df['HA_Close_EMA10'].shift(2)



# === Refactored helpers (loop-driven, replaces duplicated addtable* bodies) ===
# Refactored, loop-driven versions of the repetitive addtable* functions.
# Column names and values are designed to reproduce the originals exactly
# (regular grids are supersets; values on shared columns must match).

def _w(w):
    """2-digit-min zero padded window label: 3->'03', 600->'600'."""
    return f'{w:02d}'

def _add_shifts(df, base_col, label, shifts):
    s = df[base_col]
    for sh in shifts:
        df[f'{label}-{sh}'] = s.shift(sh)

# ----- windows / shifts ---------------------------------------------------
W_MA  = [5, 10, 15, 20, 25, 30, 40, 60, 100, 120, 200, 240, 500]
W_VOL = [5, 10, 25, 60, 100, 120, 200, 240, 600]
W_RET = [5, 10, 15, 20, 25, 60, 100, 200, 240, 600]
W_PX  = [3, 5, 10, 20, 25, 40, 60, 100, 120, 200, 240, 600]   # Min/Max/low/high
W_PCT = [3, 5, 10, 20, 25, 60, 100, 200, 600]                 # MinMaxptc / highlowptc
W_FIB = [3, 5, 10, 20, 25, 60, 100, 120, 200, 240, 600]
W_BRK = [5, 10, 20, 25, 40, 60, 100, 120, 200, 600]

PX_SHIFTS  = [1, 2, 5, 10, 20, 25, 60, 100]
VOL_SHIFTS = [1, 2, 3, 5, 10]

def _add_px_family(Stock_df, src_col, prefix, how, with_ma):
    s = Stock_df[src_col]
    for w in W_PX:
        ww = _w(w)
        base = s.rolling(window=w).agg(how)
        Stock_df[f'{prefix}{ww}'] = base
        _add_shifts(Stock_df, f'{prefix}{ww}', f'{prefix}{ww}', PX_SHIFTS)
        if with_ma:
            ma = base.rolling(window=w).mean()
            Stock_df[f'{prefix}{ww}MA']   = ma
            Stock_df[f'{prefix}{ww}MA-1'] = ma.shift(1)

def _add_pct_family(Stock_df, min_prefix, max_prefix):
    c = Stock_df['close']
    for prefix in (min_prefix, max_prefix):
        for w in W_PCT:
            ww = _w(w)
            col = f'{prefix}{ww}%'
            Stock_df[col] = ((c - Stock_df[f'{prefix}{ww}']) / c) * 100
            Stock_df[f'{col}-1'] = Stock_df[col].shift(1)
            Stock_df[f'{col}-2'] = Stock_df[col].shift(2)
    # extras present in the originals
    Stock_df[f'{min_prefix}60%-3']  = Stock_df[f'{min_prefix}60%'].shift(3)
    Stock_df[f'{min_prefix}60%dif']  = Stock_df[f'{min_prefix}60%']  - Stock_df[f'{min_prefix}60%-1']
    Stock_df[f'{min_prefix}200%dif'] = Stock_df[f'{min_prefix}200%'] - Stock_df[f'{min_prefix}200%-1']

def addtableMA(Stock_df):
    c = Stock_df['close']
    for w in W_MA:
        ww = _w(w)
        ma  = c.rolling(window=w).mean()
        std = c.rolling(window=w).std()
        Stock_df[f'MA{ww}']            = ma
        Stock_df[f'std MA{ww}']        = std
        Stock_df[f'Stock_dfGap%_c_{ww}'] = (c - ma) / c * 100
        Stock_df[f'MA/SD_{ww}']        = std / ma * 100
def addtableVol(Stock_df):
    v = Stock_df['volume']
    for w in W_VOL:
        ww = _w(w)
        Stock_df[f'VolMA{ww}']  = v.rolling(window=w).mean()
        Stock_df[f'VolMax{ww}'] = v.rolling(window=w).max()
    for sh in VOL_SHIFTS:
        Stock_df[f'volume-{sh}'] = v.shift(sh)
        for w in W_VOL:
            ww = _w(w)
            Stock_df[f'VolMax-{sh}_{ww}'] = Stock_df[f'VolMax{ww}'].shift(sh)
def addtablereturn(Stock_df):
    r = Stock_df['retune_chg']
    for w in W_RET:
        ww = _w(w)
        Stock_df[f'retune_chg_ma{ww}']    = r.rolling(window=w).mean()
        Stock_df[f'retune_chg_ma{ww}std'] = r.rolling(window=w).std()
    Stock_df['retune_chg_ma05<10<25'] = (Stock_df['retune_chg_ma05'] < Stock_df['retune_chg_ma10']) & (Stock_df['retune_chg_ma10'] < Stock_df['retune_chg_ma25'])
    Stock_df['retune_chg_ma05<10<25-1'] = Stock_df['retune_chg_ma05<10<25'].shift(1)
    Stock_df['retune_chg_ma05<10<25std'] = (Stock_df['retune_chg_ma05std'] < Stock_df['retune_chg_ma10std']) & (Stock_df['retune_chg_ma10std'] < Stock_df['retune_chg_ma25std']) & (Stock_df['retune_chg_ma25'] > 0)
    Stock_df['retune_chg_ma05<10<25std-1'] = Stock_df['retune_chg_ma05<10<25std'].shift(1)
    Stock_df['retune_chg_ma25>0'] = (Stock_df['retune_chg_ma25'] > 0)
def addtableVal(Stock_df):
  Stock_df['ValMA25'] =  Stock_df['VolMA25'] * Stock_df['close']
  Stock_df['ValMA60'] = Stock_df['VolMA60'] * Stock_df['close']
  Stock_df['ValMA120'] = Stock_df['VolMA120'] * Stock_df['close']
  Stock_df['ValMA240'] = Stock_df['VolMA240'] * Stock_df['close']
  Stock_df['ValMA600'] = Stock_df['VolMA600'] * Stock_df['close']


def addtableMin(Stock_df):
    _add_px_family(Stock_df, 'close', 'PMin', 'min', with_ma=True)
def addtableMax(Stock_df):
    _add_px_family(Stock_df, 'close', 'PMax', 'max', with_ma=True)
def addtablelow(Stock_df):
    _add_px_family(Stock_df, 'low', 'PLow', 'min', with_ma=False)
def addtablehigh(Stock_df):
    _add_px_family(Stock_df, 'high', 'PHIGH', 'max', with_ma=False)
def addtableMinMaxptc(Stock_df):
    _add_pct_family(Stock_df, 'PMin', 'PMax')
def addtablehighlowptc(Stock_df):
    _add_pct_family(Stock_df, 'PLow', 'PHIGH')
def addtablefibonacci_ptc(Stock_df):
    c = Stock_df['close']
    for w in W_FIB:
        ww = _w(w)
        rng = Stock_df[f'PMax{ww}'] - Stock_df[f'PMin{ww}']
        Stock_df[f'PRangeMaxMin{ww}'] = rng
        pct_name = f'PRangeMaxMin{ww}%%' if w == 3 else f'PRangeMaxMin{ww}%'  # keep original (typo'd) name for w=3
        Stock_df[pct_name] = (rng / Stock_df[f'PMin{ww}']) * 100
        as_of = c - Stock_df[f'PMin{ww}']
        Stock_df[f'P_as_of{ww}'] = as_of
        fibo = ((rng - as_of) / rng) * 100
        Stock_df[f'P_fibo_{ww}'] = fibo
        for sh in (1, 2, 5, 10):
            Stock_df[f'P_fibo_{ww}-{sh}'] = fibo.shift(sh)
def addtableEMA(Stock_df):
  Stock_df['EMA3'] = Stock_df['close'].ewm(span=3).mean()
  Stock_df['EMA12'] = Stock_df['close'].ewm(span=12).mean()
  Stock_df['EMA26'] = Stock_df['close'].ewm(span=26).mean()
  Stock_df['EMA40'] = Stock_df['close'].ewm(span=40).mean()
  Stock_df['EMA75'] = Stock_df['close'].ewm(span=75).mean()
  Stock_df['EMA89'] = Stock_df['close'].ewm(span=89).mean()

  Stock_df['STD_EMA3'] = Stock_df['close'].ewm(span=3).std()
  Stock_df['STD_EMA12'] = Stock_df['close'].ewm(span=12).std()
  Stock_df['STD_EMA26'] = Stock_df['close'].ewm(span=26).std()
  Stock_df['STD_EMA75'] = Stock_df['close'].ewm(span=75).std()
  Stock_df['STD_EMA89'] = Stock_df['close'].ewm(span=89).std()

  Stock_df['EMA3-1'] = Stock_df['EMA3'].shift(1)
  Stock_df['EMA12-1'] = Stock_df['EMA12'].shift(1)
  Stock_df['EMA26-1'] = Stock_df['EMA26'].shift(1)
  Stock_df['EMA75-1'] = Stock_df['EMA75'].shift(1)
  Stock_df['EMA89-1'] = Stock_df['EMA89'].shift(1)

  Stock_df['STD_EMA3-1'] = Stock_df['STD_EMA3'].shift(1)
  Stock_df['STD_EMA12-1'] = Stock_df['STD_EMA12'].shift(1)
  Stock_df['STD_EMA26-1'] = Stock_df['STD_EMA26'].shift(1)
  Stock_df['STD_EMA75-1'] = Stock_df['STD_EMA75'].shift(1)
  Stock_df['STD_EMA89-1'] = Stock_df['STD_EMA89'].shift(1)


  Stock_df['EMA12%C'] = ((Stock_df['close'] - Stock_df['EMA12'] ) / Stock_df['close']) *100
  Stock_df['EMA26%C'] = ((Stock_df['close'] - Stock_df['EMA26']  ) / Stock_df['close']) *100
  Stock_df['EMA75%C'] = ((Stock_df['close'] - Stock_df['EMA75']   ) / Stock_df['close']) *100
  Stock_df['EMA89%C'] = ((Stock_df['close'] - Stock_df['EMA89']  ) / Stock_df['close']) *100
  Stock_df['EMA12%C-1'] = Stock_df['EMA12%C'].shift(1)
  Stock_df['EMA26%C-1'] = Stock_df['EMA26%C'].shift(1)
  Stock_df['EMA75%C-1'] = Stock_df['EMA75%C'].shift(1)
  Stock_df['EMA89%C-1'] = Stock_df['EMA89%C'].shift(1)
  Stock_df['EMA12-12'] = Stock_df['EMA12'].shift(11)
  Stock_df['EMA26-26'] = Stock_df['EMA26'].shift(25)
  Stock_df['EMA75-75'] = Stock_df['EMA75'].shift(74)
  Stock_df['%EMA12-26'] = ((Stock_df['EMA12'] - Stock_df['EMA26']) / Stock_df['EMA12'])*100

  Stock_df['EMA03'] = Stock_df['close'].ewm(span=3).mean()
  Stock_df['EMA05'] = Stock_df['close'].ewm(span=5).mean()
  Stock_df['EMA10'] = Stock_df['close'].ewm(span=10).mean()
  Stock_df['EMA15'] = Stock_df['close'].ewm(span=15).mean()
  Stock_df['EMA20'] = Stock_df['close'].ewm(span=20).mean()
  Stock_df['EMA25'] = Stock_df['close'].ewm(span=25).mean()
  Stock_df['EMA30'] = Stock_df['close'].ewm(span=30).mean()
  Stock_df['EMA40'] = Stock_df['close'].ewm(span=40).mean()
  Stock_df['EMA60'] = Stock_df['close'].ewm(span=60).mean()
  Stock_df['EMA100'] = Stock_df['close'].ewm(span=100).mean()
  Stock_df['EMA120'] = Stock_df['close'].ewm(span=120).mean()
  Stock_df['EMA200'] = Stock_df['close'].ewm(span=200).mean()
  Stock_df['EMA240'] = Stock_df['close'].ewm(span=240).mean()
  Stock_df['EMA500'] = Stock_df['close'].ewm(span=500).mean()
  Stock_df['STD_EMA05'] = Stock_df['close'].ewm(span=5).std()
  Stock_df['STD_EMA10'] = Stock_df['close'].ewm(span=10).std()
  Stock_df['STD_EMA15'] = Stock_df['close'].ewm(span=15).std()
  Stock_df['STD_EMA20'] = Stock_df['close'].ewm(span=20).std()
  Stock_df['STD_EMA25'] = Stock_df['close'].ewm(span=25).std()
  Stock_df['STD_EMA30'] = Stock_df['close'].ewm(span=30).std()
  Stock_df['STD_EMA40'] = Stock_df['close'].ewm(span=40).std()
  Stock_df['STD_EMA60'] = Stock_df['close'].ewm(span=60).std()
  Stock_df['STD_EMA100'] = Stock_df['close'].ewm(span=100).std()
  Stock_df['STD_EMA120'] = Stock_df['close'].ewm(span=120).std()
  Stock_df['STD_EMA200'] = Stock_df['close'].ewm(span=200).std()
  Stock_df['STD_EMA240'] = Stock_df['close'].ewm(span=240).std()
  Stock_df['STD_EMA500'] = Stock_df['close'].ewm(span=500).std()

  Stock_df['EMA05%C']=(Stock_df['close'] - Stock_df['EMA05'] )/Stock_df['close']*100
  Stock_df['EMA10%C']=(Stock_df['close'] - Stock_df['EMA10'] )/Stock_df['close']*100
  Stock_df['EMA15%C']=(Stock_df['close'] - Stock_df['EMA15'] )/Stock_df['close']*100
  Stock_df['EMA20%C']=(Stock_df['close'] - Stock_df['EMA20'] )/Stock_df['close']*100
  Stock_df['EMA25%C']=(Stock_df['close'] - Stock_df['EMA25'] )/Stock_df['close']*100
  Stock_df['EMA30%C']=(Stock_df['close'] - Stock_df['EMA30'] )/Stock_df['close']*100
  Stock_df['EMA40%C']=(Stock_df['close'] - Stock_df['EMA40'] )/Stock_df['close']*100
  Stock_df['EMA60%C']=(Stock_df['close'] - Stock_df['EMA60'] )/Stock_df['close']*100
  Stock_df['EMA100%C']=(Stock_df['close'] - Stock_df['EMA100'] )/Stock_df['close']*100
  Stock_df['EMA120%C']=(Stock_df['close'] - Stock_df['EMA120'] )/Stock_df['close']*100
  Stock_df['EMA200%C']=(Stock_df['close'] - Stock_df['EMA200'] )/Stock_df['close']*100
  Stock_df['EMA240%C']=(Stock_df['close'] - Stock_df['EMA240'] )/Stock_df['close']*100
  Stock_df['EMA500%C']=(Stock_df['close'] -Stock_df['EMA500'])  /Stock_df['close']*100

  Stock_df['SD/EMA_05'] =Stock_df['STD_EMA05'] /Stock_df['EMA05'] *100
  Stock_df['SD/EMA_10'] =Stock_df['STD_EMA10'] /Stock_df['EMA10'] *100
  Stock_df['SD/EMA_15'] =Stock_df['STD_EMA15'] /Stock_df['EMA15'] *100
  Stock_df['SD/EMA_20'] =Stock_df['STD_EMA20'] /Stock_df['EMA20'] *100
  Stock_df['SD/EMA_25'] =Stock_df['STD_EMA25'] /Stock_df['EMA25'] *100
  Stock_df['SD/EMA_30'] =Stock_df['STD_EMA30'] /Stock_df['EMA30'] *100
  Stock_df['SD/EMA_40'] =Stock_df['STD_EMA40'] /Stock_df['EMA40'] *100
  Stock_df['SD/EMA_60'] =Stock_df['STD_EMA60'] /Stock_df['EMA60'] *100
  Stock_df['SD/EMA_100'] =Stock_df['STD_EMA100'] /Stock_df['EMA100'] *100
  Stock_df['SD/EMA_120'] =Stock_df['STD_EMA120'] /Stock_df['EMA120'] *100
  Stock_df['SD/EMA_200'] =Stock_df['STD_EMA200'] /Stock_df['EMA200'] *100
  Stock_df['SD/EMA_240'] =Stock_df['STD_EMA240'] /Stock_df['EMA240'] *100
  Stock_df['SD/EMA_500'] =Stock_df['STD_EMA500'] /Stock_df['EMA500'] *100


  Stock_df['EMA03-1']=Stock_df['EMA03'].shift(1)
  Stock_df['EMA05-1']=Stock_df['EMA05'].shift(1)
  Stock_df['EMA10-1']=Stock_df['EMA10'].shift(1)
  Stock_df['EMA12-1']=Stock_df['EMA12'].shift(1)
  Stock_df['EMA15-1']=Stock_df['EMA15'].shift(1)
  Stock_df['EMA20-1']=Stock_df['EMA20'].shift(1)
  Stock_df['EMA25-1']=Stock_df['EMA25'].shift(1)
  Stock_df['EMA26-1']=Stock_df['EMA26'].shift(1)
  Stock_df['EMA30-1']=Stock_df['EMA30'].shift(1)
  Stock_df['EMA40-1']=Stock_df['EMA40'].shift(1)
  Stock_df['EMA60-1']=Stock_df['EMA60'].shift(1)
  Stock_df['EMA75-1']=Stock_df['EMA75'].shift(1)
  Stock_df['EMA89-1']=Stock_df['EMA89'].shift(1)
  Stock_df['EMA100-1']=Stock_df['EMA100'].shift(1)
  Stock_df['EMA120-1']=Stock_df['EMA120'].shift(1)
  Stock_df['EMA200-1']=Stock_df['EMA200'].shift(1)
  Stock_df['EMA240-1']=Stock_df['EMA240'].shift(1)
  Stock_df['EMA500-1']=Stock_df['EMA500'].shift(1)
  Stock_df['STD_EMA05-1']=Stock_df['STD_EMA05'].shift(1)
  Stock_df['STD_EMA10-1']=Stock_df['STD_EMA10'].shift(1)
  Stock_df['STD_EMA15-1']=Stock_df['STD_EMA15'].shift(1)
  Stock_df['STD_EMA20-1']=Stock_df['STD_EMA20'].shift(1)
  Stock_df['STD_EMA25-1']=Stock_df['STD_EMA25'].shift(1)
  Stock_df['STD_EMA30-1']=Stock_df['STD_EMA30'].shift(1)
  Stock_df['STD_EMA40-1']=Stock_df['STD_EMA40'].shift(1)
  Stock_df['STD_EMA60-1']=Stock_df['STD_EMA60'].shift(1)
  Stock_df['STD_EMA100-1']=Stock_df['STD_EMA100'].shift(1)
  Stock_df['STD_EMA120-1']=Stock_df['STD_EMA120'].shift(1)
  Stock_df['STD_EMA200-1']=Stock_df['STD_EMA200'].shift(1)
  Stock_df['STD_EMA240-1']=Stock_df['STD_EMA240'].shift(1)
  Stock_df['STD_EMA500-1']=Stock_df['STD_EMA500'].shift(1)
  Stock_df['SD/EMA_05-1'] =Stock_df['SD/EMA_05'].shift(1)
  Stock_df['SD/EMA_10-1'] =Stock_df['SD/EMA_10'].shift(1)
  Stock_df['SD/EMA_15-1'] =Stock_df['SD/EMA_15'].shift(1)
  Stock_df['SD/EMA_20-1'] =Stock_df['SD/EMA_20'].shift(1)
  Stock_df['SD/EMA_25-1'] =Stock_df['SD/EMA_25'].shift(1)
  Stock_df['SD/EMA_30-1'] =Stock_df['SD/EMA_30'].shift(1)
  Stock_df['SD/EMA_40-1'] =Stock_df['SD/EMA_40'].shift(1)
  Stock_df['SD/EMA_60-1'] =Stock_df['SD/EMA_60'].shift(1)
  Stock_df['SD/EMA_100-1'] =Stock_df['SD/EMA_100'].shift(1)
  Stock_df['SD/EMA_120-1'] =Stock_df['SD/EMA_120'].shift(1)
  Stock_df['SD/EMA_200-1'] =Stock_df['SD/EMA_200'].shift(1)
  Stock_df['SD/EMA_240-1'] =Stock_df['SD/EMA_240'].shift(1)
  Stock_df['SD/EMA_500-1'] =Stock_df['SD/EMA_500'].shift(1)

  Stock_df['retune_chg_EMA05'] = Stock_df['retune_chg'].ewm(span=5).mean()
  Stock_df['retune_chg_EMA10'] = Stock_df['retune_chg'].ewm(span=10).mean()
  Stock_df['retune_chg_EMA15'] = Stock_df['retune_chg'].ewm(span=15).mean()
  Stock_df['retune_chg_EMA20'] = Stock_df['retune_chg'].ewm(span=20).mean()
  Stock_df['retune_chg_EMA25'] = Stock_df['retune_chg'].ewm(span=25).mean()
  Stock_df['retune_chg_EMA30'] = Stock_df['retune_chg'].ewm(span=30).mean()
  Stock_df['retune_chg_EMA40'] = Stock_df['retune_chg'].ewm(span=40).mean()
  Stock_df['retune_chg_EMA60'] = Stock_df['retune_chg'].ewm(span=60).mean()
  Stock_df['retune_chg_EMA100'] = Stock_df['retune_chg'].ewm(span=100).mean()
  Stock_df['retune_chg_EMA120'] = Stock_df['retune_chg'].ewm(span=120).mean()
  Stock_df['retune_chg_EMA200'] = Stock_df['retune_chg'].ewm(span=200).mean()
  Stock_df['retune_chg_EMA240'] = Stock_df['retune_chg'].ewm(span=240).mean()
  Stock_df['retune_chg_EMA500'] = Stock_df['retune_chg'].ewm(span=500).mean()
  Stock_df['retune_chg_STD_EMA_EMA05'] = Stock_df['retune_chg'].ewm(span=5).std()
  Stock_df['retune_chg_STD_EMA_EMA10'] = Stock_df['retune_chg'].ewm(span=10).std()
  Stock_df['retune_chg_STD_EMA_EMA15'] = Stock_df['retune_chg'].ewm(span=15).std()
  Stock_df['retune_chg_STD_EMA_EMA20'] = Stock_df['retune_chg'].ewm(span=20).std()
  Stock_df['retune_chg_STD_EMA_EMA25'] = Stock_df['retune_chg'].ewm(span=25).std()
  Stock_df['retune_chg_STD_EMA_EMA30'] = Stock_df['retune_chg'].ewm(span=30).std()
  Stock_df['retune_chg_STD_EMA_EMA40'] = Stock_df['retune_chg'].ewm(span=40).std()
  Stock_df['retune_chg_STD_EMA_EMA60'] = Stock_df['retune_chg'].ewm(span=60).std()
  Stock_df['retune_chg_STD_EMA_EMA100'] = Stock_df['retune_chg'].ewm(span=100).std()
  Stock_df['retune_chg_STD_EMA_EMA120'] = Stock_df['retune_chg'].ewm(span=120).std()
  Stock_df['retune_chg_STD_EMA_EMA200'] = Stock_df['retune_chg'].ewm(span=200).std()
  Stock_df['retune_chg_STD_EMA_EMA240'] = Stock_df['retune_chg'].ewm(span=240).std()
  Stock_df['retune_chg_STD_EMA_EMA500'] = Stock_df['retune_chg'].ewm(span=500).std()

  Stock_df['retune_chg_SD/EMA_05'] =Stock_df['retune_chg_STD_EMA_EMA05'] /Stock_df['retune_chg_EMA05'] *100
  Stock_df['retune_chg_SD/EMA_10'] =Stock_df['retune_chg_STD_EMA_EMA10'] /Stock_df['retune_chg_EMA10'] *100
  Stock_df['retune_chg_SD/EMA_15'] =Stock_df['retune_chg_STD_EMA_EMA15'] /Stock_df['retune_chg_EMA15'] *100
  Stock_df['retune_chg_SD/EMA_20'] =Stock_df['retune_chg_STD_EMA_EMA20'] /Stock_df['retune_chg_EMA20'] *100
  Stock_df['retune_chg_SD/EMA_25'] =Stock_df['retune_chg_STD_EMA_EMA25'] /Stock_df['retune_chg_EMA25'] *100
  Stock_df['retune_chg_SD/EMA_30'] =Stock_df['retune_chg_STD_EMA_EMA30'] /Stock_df['retune_chg_EMA30'] *100
  Stock_df['retune_chg_SD/EMA_40'] =Stock_df['retune_chg_STD_EMA_EMA40'] /Stock_df['retune_chg_EMA40'] *100
  Stock_df['retune_chg_SD/EMA_60'] =Stock_df['retune_chg_STD_EMA_EMA60'] /Stock_df['retune_chg_EMA60'] *100
  Stock_df['retune_chg_SD/EMA_100'] =Stock_df['retune_chg_STD_EMA_EMA100'] /Stock_df['retune_chg_EMA100'] *100
  Stock_df['retune_chg_SD/EMA_120'] =Stock_df['retune_chg_STD_EMA_EMA120'] /Stock_df['retune_chg_EMA120'] *100
  Stock_df['retune_chg_SD/EMA_200'] =Stock_df['retune_chg_STD_EMA_EMA200'] /Stock_df['retune_chg_EMA200'] *100
  Stock_df['retune_chg_SD/EMA_240'] =Stock_df['retune_chg_STD_EMA_EMA240'] /Stock_df['retune_chg_EMA240'] *100
  Stock_df['retune_chg_SD/EMA_500'] =Stock_df['retune_chg_STD_EMA_EMA500'] /Stock_df['retune_chg_EMA500'] *100

  Stock_df['EMA05-2']=Stock_df['EMA05'].shift(2)
  Stock_df['EMA10-2']=Stock_df['EMA10'].shift(2)
  Stock_df['EMA12-2']=Stock_df['EMA12'].shift(2)
  Stock_df['EMA26-2']=Stock_df['EMA25'].shift(2)
  Stock_df['EMA40-2']=Stock_df['EMA40'].shift(2)
  Stock_df['EMA60-2']=Stock_df['EMA60'].shift(2)
  Stock_df['EMA75-2']=Stock_df['EMA75'].shift(2)
  Stock_df['EMA89-2']=Stock_df['EMA89'].shift(2)
  Stock_df['EMA100-2']=Stock_df['EMA100'].shift(2)
  Stock_df['EMA200-2']=Stock_df['EMA200'].shift(2)

  Stock_df['EMA05-5']=Stock_df['EMA05'].shift(5)
  Stock_df['EMA10-5']=Stock_df['EMA10'].shift(5)
  Stock_df['EMA12-5']=Stock_df['EMA12'].shift(5)
  Stock_df['EMA26-5']=Stock_df['EMA25'].shift(5)
  Stock_df['EMA40-5']=Stock_df['EMA40'].shift(5)
  Stock_df['EMA60-5']=Stock_df['EMA60'].shift(5)
  Stock_df['EMA75-5']=Stock_df['EMA75'].shift(5)
  Stock_df['EMA89-5']=Stock_df['EMA89'].shift(5)
  Stock_df['EMA100-5']=Stock_df['EMA100'].shift(5)
  Stock_df['EMA200-5']=Stock_df['EMA200'].shift(5)

  Stock_df['EMA05-6']=Stock_df['EMA05'].shift(6)
  Stock_df['EMA10-6']=Stock_df['EMA10'].shift(6)
  Stock_df['EMA12-6']=Stock_df['EMA12'].shift(6)
  Stock_df['EMA26-6']=Stock_df['EMA25'].shift(6)
  Stock_df['EMA40-6']=Stock_df['EMA40'].shift(6)
  Stock_df['EMA60-6']=Stock_df['EMA60'].shift(6)
  Stock_df['EMA75-6']=Stock_df['EMA75'].shift(6)
  Stock_df['EMA89-6']=Stock_df['EMA89'].shift(6)
  Stock_df['EMA100-6']=Stock_df['EMA100'].shift(6)
  Stock_df['EMA200-6']=Stock_df['EMA200'].shift(6)

  Stock_df['EMA05-10']=Stock_df['EMA05'].shift(10)
  Stock_df['EMA10-10']=Stock_df['EMA10'].shift(10)
  Stock_df['EMA12-10']=Stock_df['EMA12'].shift(10)
  Stock_df['EMA26-10']=Stock_df['EMA25'].shift(10)
  Stock_df['EMA40-10']=Stock_df['EMA40'].shift(10)
  Stock_df['EMA60-10']=Stock_df['EMA60'].shift(10)
  Stock_df['EMA75-10']=Stock_df['EMA75'].shift(10)
  Stock_df['EMA89-10']=Stock_df['EMA89'].shift(10)
  Stock_df['EMA100-10']=Stock_df['EMA100'].shift(10)
  Stock_df['EMA200-10']=Stock_df['EMA200'].shift(10)

  Stock_df['EMA05-11']=Stock_df['EMA05'].shift(11)
  Stock_df['EMA10-11']=Stock_df['EMA10'].shift(11)
  Stock_df['EMA12-11']=Stock_df['EMA12'].shift(11)
  Stock_df['EMA26-11']=Stock_df['EMA25'].shift(11)
  Stock_df['EMA40-11']=Stock_df['EMA40'].shift(11)
  Stock_df['EMA60-11']=Stock_df['EMA60'].shift(11)
  Stock_df['EMA75-11']=Stock_df['EMA75'].shift(11)
  Stock_df['EMA89-11']=Stock_df['EMA89'].shift(11)
  Stock_df['EMA100-11']=Stock_df['EMA100'].shift(11)
  Stock_df['EMA200-11']=Stock_df['EMA200'].shift(11)

  Stock_df['EMA05-40']=Stock_df['EMA05'].shift(40)
  Stock_df['EMA10-40']=Stock_df['EMA10'].shift(40)
  Stock_df['EMA12-40']=Stock_df['EMA12'].shift(40)
  Stock_df['EMA26-40']=Stock_df['EMA25'].shift(40)
  Stock_df['EMA40-40']=Stock_df['EMA40'].shift(40)
  Stock_df['EMA60-40']=Stock_df['EMA60'].shift(40)
  Stock_df['EMA75-40']=Stock_df['EMA75'].shift(40)
  Stock_df['EMA89-40']=Stock_df['EMA89'].shift(40)
  Stock_df['EMA100-40']=Stock_df['EMA100'].shift(40)
  Stock_df['EMA200-40']=Stock_df['EMA200'].shift(40)

  Stock_df['EMA05-75']=Stock_df['EMA05'].shift(75)
  Stock_df['EMA10-75']=Stock_df['EMA10'].shift(75)
  Stock_df['EMA12-75']=Stock_df['EMA12'].shift(75)
  Stock_df['EMA26-75']=Stock_df['EMA25'].shift(75)
  Stock_df['EMA26-40']=Stock_df['EMA40'].shift(75)
  Stock_df['EMA60-75']=Stock_df['EMA60'].shift(75)
  Stock_df['EMA75-75']=Stock_df['EMA75'].shift(75)
  Stock_df['EMA89-75']=Stock_df['EMA89'].shift(75)
  Stock_df['EMA100-75']=Stock_df['EMA100'].shift(75)
  Stock_df['EMA200-75']=Stock_df['EMA200'].shift(75)


def addtablebreakHL(Stock_df):
    c = Stock_df['close']
    for w in W_BRK:
        ww = _w(w)
        bmax = c > Stock_df[f'PMax{ww}-1']
        bmin = c < Stock_df[f'PMin{ww}-1']
        Stock_df[f'BreakPMax{ww}'] = bmax
        Stock_df[f'BreakPMin{ww}'] = bmin
        for sh in (1, 2, 5):
            Stock_df[f'BreakPMax{ww}-{sh}'] = bmax.shift(sh)
            Stock_df[f'BreakPMin{ww}-{sh}'] = bmin.shift(sh)
def addtableCal(Stock_df):
  Stock_df['closeMax25'] = Stock_df['close'].rolling(window=25).max()
  Stock_df['%closeMax25'] = ((Stock_df['closeMax25'] - Stock_df['close']) / Stock_df['close'])*100
  Stock_df['Maxhigh25'] = Stock_df['high'].rolling(window=25).max()
  Stock_df['Maxhigh25-1'] = Stock_df['Maxhigh25'].shift(1)
  Stock_df['MA25-25'] = Stock_df['MA25'].shift(25)
  Stock_df['SlopeMA25'] = (Stock_df['MA25'] - Stock_df['MA25-25']) / 25

def addtableCalinfo(Stock_df):
  Stock_df['%VolumeTurnoveFloat'] = (Stock_df['volume']/Stock_df['floatShares']) * 100
  Stock_df['%VolumeTurnove'] = (Stock_df['volume']/Stock_df['sharesOutstanding']) *100
  Stock_df['%Mcapturnover'] = (Stock_df['Values']/Stock_df['Mcap'])*100

def addtableADX(Stock_df):
  Stock_df['ADX'] = ta.trend.adx(Stock_df['high'],Stock_df['low'],Stock_df["close"],14,False)
  Stock_df['DI+'] = ta.trend.adx_pos(Stock_df['high'],Stock_df['low'],Stock_df["close"],14,False)
  Stock_df['DI-'] = ta.trend.adx_neg(Stock_df['high'],Stock_df['low'],Stock_df["close"],14,False)
  Stock_df['ADX_Hline1'] = 30
  Stock_df['ADX_Hline2'] = 45
  Stock_df['ADX-1'] = Stock_df['ADX'].shift(1)
  Stock_df['DI+-1'] = Stock_df['DI+'].shift(1)
  Stock_df['DI--1']= Stock_df['DI-'].shift(1)
  Stock_df['ADX-2'] =Stock_df['ADX'].shift(2)
  Stock_df['DI+-2'] = Stock_df['DI+'].shift(2)
  Stock_df['DI--2'] = Stock_df['DI-'].shift(2)



  # Stock_df['EMA12-1'] = Stock_df['close'].ewm(span=12).mean().shift(1)
  # Stock_df['EMA26-1'] = Stock_df['close'].ewm(span=26).mean().shift(1)
  # Stock_df['EMA75-1'] = Stock_df['close'].ewm(span=75).mean().shift(1)

  # for bb

def addtable1_bb(Stock_df):
  Stock_df['rstd'] = Stock_df['close'].rolling(window=25).std()
  Stock_df['upper_band'] = Stock_df['MA25'] + 2 * Stock_df['rstd']
  Stock_df['lower_band'] = Stock_df['MA25'] - 2 * Stock_df['rstd']
  Stock_df['upper_band_1'] = Stock_df['upper_band'].shift(1)
  Stock_df['lower_band_1'] = Stock_df['lower_band'].shift(1)

def addtable1_donc_c(Stock_df):
  Stock_df['hi'] = Stock_df['high']
  Stock_df['lo'] = Stock_df['low']
  Stock_df['uc'] = Stock_df['hi'].rolling(n, min_periods=n).max()
  Stock_df['lc'] = Stock_df['lo'].rolling(n, min_periods=n).min()
  Stock_df['mc'] = (Stock_df['uc']+Stock_df['lc'])/2
  Stock_df['uc-1'] = Stock_df['uc'].shift(1)
  Stock_df['lc-1'] = Stock_df['lc'].shift(1)
  Stock_df['MA25'] = Stock_df['close'].rolling(window=25).mean()

def addtable1_donc_25(Stock_df):
  Stock_df['hi25'] = Stock_df['high']
  Stock_df['lo25'] = Stock_df['low']
  Stock_df['uc25'] = Stock_df['hi'].rolling(25, min_periods=n).max()
  Stock_df['lc25'] = Stock_df['lo'].rolling(25, min_periods=n).min()
  Stock_df['mc25'] = (Stock_df['uc']+Stock_df['lc'])/2
  Stock_df['uc-1_25'] = Stock_df['uc'].shift(25)
  Stock_df['lc-1_25'] = Stock_df['lc'].shift(25)
  # Stock_df['MA60'] = Stock_df['close'].rolling(window=25).mean()

def addtable1_donc_60(Stock_df):
  Stock_df['hi60'] = Stock_df['high']
  Stock_df['lo60'] = Stock_df['low']
  Stock_df['uc60'] = Stock_df['hi'].rolling(60, min_periods=n).max()
  Stock_df['lc60'] = Stock_df['lo'].rolling(60, min_periods=n).min()
  Stock_df['mc60'] = (Stock_df['uc']+Stock_df['lc'])/2
  Stock_df['uc-1_60'] = Stock_df['uc'].shift(60)
  Stock_df['lc-1_60'] = Stock_df['lc'].shift(60)
  # Stock_df['MA200'] = Stock_df['close'].rolling(window=60).mean()

def addtable1_donc_200(Stock_df):
  Stock_df['hi200'] = Stock_df['high']
  Stock_df['lo200'] = Stock_df['low']
  Stock_df['uc200'] = Stock_df['hi'].rolling(200, min_periods=n).max()
  Stock_df['lc200'] = Stock_df['lo'].rolling(200, min_periods=n).min()
  Stock_df['mc200'] = (Stock_df['uc']+Stock_df['lc'])/2
  Stock_df['uc-1_200'] = Stock_df['uc'].shift(200)
  Stock_df['lc-1_200'] = Stock_df['lc'].shift(200)
  # Stock_df['MA200'] = Stock_df['close'].rolling(window=200).mean()

def addtable1_rsi(Stock_df):
  Stock_df['RSI14'] = ta.momentum.rsi(Stock_df["close"], 14,  False)
  Stock_df['RSI14_last'] = ta.momentum.rsi(Stock_df["close"].shift(1), 14,  False)
  Stock_df['RSI14-1'] = Stock_df['RSI14'].shift(1)
  Stock_df['RSI14-2'] = Stock_df['RSI14'].shift(2)
  Stock_df['RSI14-3'] = Stock_df['RSI14'].shift(3)
  Stock_df['RSI14-4'] = Stock_df['RSI14'].shift(4)
  Stock_df['RSI14-5'] = Stock_df['RSI14'].shift(5)
  Stock_df['RSI14-10'] = Stock_df['RSI14'].shift(10)
  Stock_df['RSI14-12'] = Stock_df['RSI14'].shift(12)
  Stock_df['RSI14-14'] = Stock_df['RSI14'].shift(14)
  Stock_df['RSI14-26'] = Stock_df['RSI14'].shift(26)
  Stock_df['RSI14-40'] = Stock_df['RSI14'].shift(40)
  Stock_df['RSI14-75'] = Stock_df['RSI14'].shift(75)
  Stock_df['RSI14-89'] = Stock_df['RSI14'].shift(89)
  Stock_df['RSI25'] = ta.momentum.rsi(Stock_df["close"], 25,  False)
  Stock_df['RSI25_last'] = ta.momentum.rsi(Stock_df["close"].shift(1), 25,  False)
  Stock_df['RSI60'] = ta.momentum.rsi(Stock_df["close"], 60,  False)
  Stock_df['RSI60_last'] = ta.momentum.rsi(Stock_df["close"].shift(1), 60,  False)

def addtable1_MCDX(Stock_df):
  Stock_df['RSI50'] = ta.momentum.rsi(Stock_df["close"], 50,  False)
  Stock_df['MCDX_red'] = (Stock_df['RSI50'] - 50) * 1.5
  Stock_df['RSI30'] = ta.momentum.rsi(Stock_df["close"], 50,  False)
  Stock_df['MCDX_yellow'] = (Stock_df['RSI30'] - 30) * 0.7
  Stock_df['MCDX_green'] = 20 - (Stock_df['MCDX_red']+Stock_df['MCDX_yellow'])
  Stock_df['MCDX_red-1'] = Stock_df['MCDX_red'].shift(1)
  Stock_df['MCDX_yellow-1'] = Stock_df['MCDX_yellow'].shift(1)
  Stock_df['MCDX_green-1'] = Stock_df['MCDX_green'].shift(1)
  Stock_df['MCDX_red-2'] = Stock_df['MCDX_red'].shift(2)
  Stock_df['MCDX_yellow-2'] = Stock_df['MCDX_yellow'].shift(2)
  Stock_df['MCDX_green-2'] = Stock_df['MCDX_green'].shift(2)
  Stock_df['MCDX_red-5'] = Stock_df['MCDX_red'].shift(5)
  Stock_df['MCDX_yellow-5'] = Stock_df['MCDX_yellow'].shift(5)
  Stock_df['MCDX_green-5'] = Stock_df['MCDX_green'].shift(5)
  Stock_df['MCDX_max_red'] = Stock_df['MCDX_red'].rolling(window=10).max()
  Stock_df['MCDX_max_yellow'] = Stock_df['MCDX_yellow'].rolling(window=10).max()
  Stock_df['MCDX_max_green'] = Stock_df['MCDX_green'].rolling(window=10).max()


def addtable_vp600(Stock_df):
  grouped600 = Stock_df.groupby("close")
  dfgroup600 = grouped600[["volume"]].aggregate(np.sum)
  dfgroup600 = dfgroup600.sort_values(by='volume')
  dfgroup600['close_vp600'] = dfgroup600.index
  last_row_vp600 = dfgroup600.tail(1)
  last_row_vp600 = last_row_vp600[['close_vp600']]
  vp600 = last_row_vp600.iloc[[0],[0]]

  for i in last_row_vp600['close_vp600']:
      Stock_df['VP600']  = i

def addtable_vp200(Stock_df):
  Stock_df200 = Stock_df.tail(200)
  grouped200 = Stock_df200.groupby("close")
  dfgroup200 = grouped200[["volume"]].aggregate(np.sum)
  dfgroup200 = dfgroup200.sort_values(by='volume')
  dfgroup200['close_vp200'] = dfgroup200.index
  last_row_vp200 = dfgroup200.tail(1)
  last_row_vp200 = last_row_vp200[['close_vp200']]
  vp200 = last_row_vp200.iloc[[0],[0]]

  for i in last_row_vp200['close_vp200']:
      Stock_df['VP200']  = i

def addtable_vp60(Stock_df):
  Stock_df60 = Stock_df.tail(60)
  grouped60 = Stock_df60.groupby("close")
  dfgroup60 = grouped60[["volume"]].aggregate(np.sum)
  dfgroup60 = dfgroup60.sort_values(by='volume')
  dfgroup60['close_vp60'] = dfgroup60.index
  last_row_vp60 = dfgroup60.tail(1)
  last_row_vp60 = last_row_vp60[['close_vp60']]
  vp60 = last_row_vp60.iloc[[0],[0]]

  for i in last_row_vp60['close_vp60']:
      Stock_df['VP60']  = i

def addtable_vp25(Stock_df):
  Stock_df25 = Stock_df.tail(25)
  grouped25 = Stock_df25.groupby("close")
  dfgroup25 = grouped25[["volume"]].aggregate(np.sum)
  dfgroup25 = dfgroup25.sort_values(by='volume')
  dfgroup25['close_vp25'] = dfgroup25.index
  last_row_vp25 = dfgroup25.tail(1)
  last_row_vp25 = last_row_vp25[['close_vp25']]
  vp25 = last_row_vp25.iloc[[0],[0]]

  for i in last_row_vp25['close_vp25']:
      Stock_df['VP25']  = i


def addtable_downtrend(Stock_df):
  Stock_df['MA05_20TF'] = Stock_df['MA05'] > Stock_df['MA20']
  Stock_df['MA20_25TF'] = Stock_df['MA20'] > Stock_df['MA25']
  Stock_df['MA25_60TF'] = Stock_df['MA25'] > Stock_df['MA60']
  Stock_df['MA60_100TF'] = Stock_df['MA60'] > Stock_df['MA100']
  Stock_df['MA100_200TF'] = Stock_df['MA100'] > Stock_df['MA200']

def addtable_Volatility(Stock_df):
  Stock_df['Volatility%_5d'] = Stock_df['retune_log'].rolling(window=5).std()
  Stock_df['Volatility%_10d'] = Stock_df['retune_log'].rolling(window=10).std()
  Stock_df['Volatility%_20d'] = Stock_df['retune_log'].rolling(window=20).std()
  Stock_df['Volatility%_25d'] = Stock_df['retune_log'].rolling(window=25).std()
  Stock_df['Volatility%_40d'] = Stock_df['retune_log'].rolling(window=40).std()
  Stock_df['Volatility%_60d'] = Stock_df['retune_log'].rolling(window=60).std()
  Stock_df['Volatility%_100d'] = Stock_df['retune_log'].rolling(window=100).std()
  Stock_df['Volatility%_120d'] = Stock_df['retune_log'].rolling(window=120).std()
  Stock_df['Volatility%_200d'] = Stock_df['retune_log'].rolling(window=200).std()
  Stock_df['Volatility%_240d'] = Stock_df['retune_log'].rolling(window=240).std()
  Stock_df['Volatility%_5d-1'] = Stock_df['Volatility%_5d'].shift(1)
  Stock_df['Volatility%_10d-1'] = Stock_df['Volatility%_10d'].shift(1)
  Stock_df['Volatility%_20d-1'] = Stock_df['Volatility%_20d'].shift(1)
  Stock_df['Volatility%_25d-1'] = Stock_df['Volatility%_25d'].shift(1)
  Stock_df['Volatility%_40d-1'] = Stock_df['Volatility%_40d'].shift(1)
  Stock_df['Volatility%_60d-1'] = Stock_df['Volatility%_60d'].shift(1)
  Stock_df['Volatility%_100d-1'] = Stock_df['Volatility%_100d'].shift(1)
  Stock_df['Volatility%_120d-1'] = Stock_df['Volatility%_120d'].shift(1)
  Stock_df['Volatility%_200d-1'] = Stock_df['Volatility%_200d'].shift(1)
  Stock_df['Volatility%_240d-1'] = Stock_df['Volatility%_240d'].shift(1)
  Stock_df['Volatility%_5d-2'] = Stock_df['Volatility%_5d'].shift(2)
  Stock_df['Volatility%_10d-2'] = Stock_df['Volatility%_10d'].shift(2)
  Stock_df['Volatility%_20d-2'] = Stock_df['Volatility%_20d'].shift(2)
  Stock_df['Volatility%_25d-2'] = Stock_df['Volatility%_25d'].shift(2)
  Stock_df['Volatility%_40d-2'] = Stock_df['Volatility%_40d'].shift(2)
  Stock_df['Volatility%_60d-2'] = Stock_df['Volatility%_60d'].shift(2)
  Stock_df['Volatility%_100d-2'] = Stock_df['Volatility%_100d'].shift(2)
  Stock_df['Volatility%_120d-2'] = Stock_df['Volatility%_120d'].shift(2)
  Stock_df['Volatility%_200d-2'] = Stock_df['Volatility%_200d'].shift(2)
  Stock_df['Volatility%_240d-2'] = Stock_df['Volatility%_240d'].shift(2)
  Stock_df['Volatility%_5d-5'] = Stock_df['Volatility%_5d'].shift(5)
  Stock_df['Volatility%_10d-5'] = Stock_df['Volatility%_10d'].shift(5)
  Stock_df['Volatility%_20d-5'] = Stock_df['Volatility%_20d'].shift(5)
  Stock_df['Volatility%_25d-5'] = Stock_df['Volatility%_25d'].shift(5)
  Stock_df['Volatility%_40d-5'] = Stock_df['Volatility%_40d'].shift(5)
  Stock_df['Volatility%_60d-5'] = Stock_df['Volatility%_60d'].shift(5)
  Stock_df['Volatility%_100d-5'] = Stock_df['Volatility%_100d'].shift(5)
  Stock_df['Volatility%_120d-5'] = Stock_df['Volatility%_120d'].shift(5)
  Stock_df['Volatility%_200d-5'] = Stock_df['Volatility%_200d'].shift(5)
  Stock_df['Volatility%_240d-5'] = Stock_df['Volatility%_240d'].shift(5)
  Stock_df['Volatility%_5d-10'] = Stock_df['Volatility%_5d'].shift(10)
  Stock_df['Volatility%_10d-10'] = Stock_df['Volatility%_10d'].shift(10)
  Stock_df['Volatility%_20d-10'] = Stock_df['Volatility%_20d'].shift(10)
  Stock_df['Volatility%_25d-10'] = Stock_df['Volatility%_25d'].shift(10)
  Stock_df['Volatility%_40d-10'] = Stock_df['Volatility%_40d'].shift(10)
  Stock_df['Volatility%_60d-10'] = Stock_df['Volatility%_60d'].shift(10)
  Stock_df['Volatility%_100d-10'] = Stock_df['Volatility%_100d'].shift(10)
  Stock_df['Volatility%_120d-10'] = Stock_df['Volatility%_120d'].shift(10)
  Stock_df['Volatility%_200d-10'] = Stock_df['Volatility%_200d'].shift(10)
  Stock_df['Volatility%_240d-10'] = Stock_df['Volatility%_240d'].shift(10)
  Stock_df['Volatility%_5d-25'] = Stock_df['Volatility%_5d'].shift(25)
  Stock_df['Volatility%_10d-25'] = Stock_df['Volatility%_10d'].shift(25)
  Stock_df['Volatility%_20d-25'] = Stock_df['Volatility%_20d'].shift(25)
  Stock_df['Volatility%_25d-25'] = Stock_df['Volatility%_25d'].shift(25)
  Stock_df['Volatility%_40d-25'] = Stock_df['Volatility%_40d'].shift(25)
  Stock_df['Volatility%_60d-25'] = Stock_df['Volatility%_60d'].shift(25)
  Stock_df['Volatility%_100d-25'] = Stock_df['Volatility%_100d'].shift(25)
  Stock_df['Volatility%_120d-25'] = Stock_df['Volatility%_120d'].shift(25)
  Stock_df['Volatility%_200d-25'] = Stock_df['Volatility%_200d'].shift(25)
  Stock_df['Volatility%_240d-25'] = Stock_df['Volatility%_240d'].shift(25)




def addtable1_rsiweekly(StockW_df):
  StockW_df['RSI14W'] = ta.momentum.rsi(StockW_df["close"], 14,  False)
  StockW_df['RSI14W-1'] = StockW_df['RSI14W'].shift(1)
  StockW_df['RSI14W-2'] = StockW_df['RSI14W'].shift(2)
  StockW_df['RSI14W-3'] = StockW_df['RSI14W'].shift(3)
  StockW_df['RSI14W-4'] = StockW_df['RSI14W'].shift(4)
  StockW_df['RSI14W-5'] = StockW_df['RSI14W'].shift(5)
  StockW_df['RSI14W-10'] = StockW_df['RSI14W'].shift(10)
  StockW_df['RSI14W-12'] = StockW_df['RSI14W'].shift(12)
  StockW_df['RSI14W-14'] = StockW_df['RSI14W'].shift(14)
  StockW_df['RSI14W-26'] = StockW_df['RSI14W'].shift(26)
  StockW_df['RSI14W-40'] = StockW_df['RSI14W'].shift(40)
  StockW_df['RSI14W-75'] = StockW_df['RSI14W'].shift(75)
  StockW_df['RSI14W-89'] = StockW_df['RSI14W'].shift(89)
  # StockW_df['RSI25'] = ta.momentum.rsi(StockW_df["close"], 25,  False)
  # StockW_df['RSI25_last'] = ta.momentum.rsi(StockW_df["close"].shift(1), 25,  False)
  # StockW_df['RSI60'] = ta.momentum.rsi(StockW_df["close"], 60,  False)
  # StockW_df['RSI60_last'] = ta.momentum.rsi(StockW_df["close"].shift(1), 60,  False)


# add condition in col table 1st time
def addlaststocktable(StockL_df):
  StockL_df['2000THB'] = 2000/StockL_df['close']
  StockL_df['4000THB'] = 4000/StockL_df['close']
  StockL_df['check_closeMax25'] = (StockL_df['close'] == StockL_df['closeMax25'])
  StockL_df['%check_closeMax25'] = ((StockL_df['close'] - StockL_df['closeMax25'])/ StockL_df['close']) * 100
  StockL_df['BreakMA5'] = (StockL_df['close'] > StockL_df['MA05']) & (StockL_df['close-1'] < StockL_df['MA05'])
  StockL_df['BreakMA10']=(StockL_df['close']>StockL_df['MA10'])&(StockL_df['close-1']<StockL_df['MA10'])
  StockL_df['BreakMA20']=(StockL_df['close']>StockL_df['MA20'])&(StockL_df['close-1']<StockL_df['MA20'])
  StockL_df['BreakMA25']=(StockL_df['close']>StockL_df['MA25'])&(StockL_df['close-1']<StockL_df['MA25'])
  StockL_df['BreakMA30']=(StockL_df['close']>StockL_df['MA30'])&(StockL_df['close-1']<StockL_df['MA30'])
  StockL_df['BreakMA40']=(StockL_df['close']>StockL_df['MA40'])&(StockL_df['close-1']<StockL_df['MA40'])
  StockL_df['BreakMA60']=(StockL_df['close']>StockL_df['MA60'])&(StockL_df['close-1']<StockL_df['MA60'])
  StockL_df['BreakMA120']=(StockL_df['close']>StockL_df['MA120'])&(StockL_df['close-1']<StockL_df['MA120'])
  StockL_df['BreakMA240']=(StockL_df['close']>StockL_df['MA240'])&(StockL_df['close-1']<StockL_df['MA240'])
  StockL_df['BreakMA500']=(StockL_df['close']>StockL_df['MA500'])&(StockL_df['close-1']<StockL_df['MA500'])

  StockL_df['BreakMaxclose5'] = (StockL_df['close'] > StockL_df['PMax05-1']) & (StockL_df['close-1'] < StockL_df['PMax05-1'])
  StockL_df['BreakMaxclose10']=(StockL_df['close']>StockL_df['PMax10-1'])&(StockL_df['close-1']<StockL_df['PMax10-1'])
  StockL_df['BreakMaxclose20']=(StockL_df['close']>StockL_df['PMax20-1'])&(StockL_df['close-1']<StockL_df['PMax20-1'])
  StockL_df['BreakMaxclose25']=(StockL_df['close']>StockL_df['PMax25-1'])&(StockL_df['close-1']<StockL_df['PMax25-1'])
  StockL_df['BreakMaxclose60']=(StockL_df['close']>StockL_df['PMax60-1'])&(StockL_df['close-1']<StockL_df['PMax60-1'])
  StockL_df['BreakMaxclose100']=(StockL_df['close']>StockL_df['PMax100-1'])&(StockL_df['close-1']<StockL_df['PMax100-1'])
  StockL_df['BreakMaxclose120']=(StockL_df['close']>StockL_df['PMax120-1'])&(StockL_df['close-1']<StockL_df['PMax120-1'])
  StockL_df['BreakMaxclose240']=(StockL_df['close']>StockL_df['PMax240-1'])&(StockL_df['close-1']<StockL_df['PMax240-1'])
  StockL_df['BreakMaxclose600']=(StockL_df['close']>StockL_df['PMax600-1'])&(StockL_df['close-1']<StockL_df['PMax600-1'])

  StockL_df['VolMA60x3'] = (StockL_df['VolMA60'] ) * 3
  StockL_df['VolMA60x2'] = (StockL_df['VolMA60'] ) * 2
  StockL_df['OverVolMAx3'] = (StockL_df['volume'] > StockL_df['VolMA60x3'] )
  StockL_df['OverVolMAx2'] = (StockL_df['volume'] > StockL_df['VolMA60x2'] )
  StockL_df['%chg'] = ((StockL_df['close'] - StockL_df['close-1'])/StockL_df['close-1']) *100
  StockL_df['%chg-1'] = ((StockL_df['close-1'] - StockL_df['close-2'])/StockL_df['close-2']) *100
  StockL_df['%chginday'] = ((StockL_df['close'] - StockL_df['open'])/StockL_df['open']) *100
  StockL_df['Break_UC'] = (StockL_df['uc-1'] < StockL_df['uc'])
  StockL_df['Break_LC'] = (StockL_df['lc-1'] > StockL_df['lc'])
  # StockL_df['vp60%'] = (StockL_df['close'] - StockL_df['VP60']) / StockL_df['VP60']
  # StockL_df['vp200%'] = (StockL_df['close'] - StockL_df['VP200']) / StockL_df['VP200']
  # StockL_df['vp600%'] = (StockL_df['close'] - StockL_df['VP600']) / StockL_df['VP600']
  StockL_df['RSI_up'] = (StockL_df['RSI14'] > StockL_df['RSI25'])&(StockL_df['RSI14'] > StockL_df['RSI60'])&(StockL_df['RSI25'] > StockL_df['RSI60'])

  StockL_df['RSI14_thoughtsback'] = (StockL_df['RSI14'] < 70) & ((StockL_df['RSI14-1'] > 70) | (StockL_df['RSI14-2'] > 70) | (StockL_df['RSI14-3'] > 70) | (StockL_df['RSI14-4'] > 70))

  StockL_df['ZoneUpper_fibo_03']=( StockL_df['P_fibo_03']<= 20)
  StockL_df['ZoneUpper_fibo_05']=( StockL_df['P_fibo_05']<= 20)
  StockL_df['ZoneUpper_fibo_10']=( StockL_df['P_fibo_10']<= 20)
  StockL_df['ZoneUpper_fibo_20']=( StockL_df['P_fibo_20']<= 20)
  StockL_df['ZoneUpper_fibo_25']=( StockL_df['P_fibo_25']<= 20)
  StockL_df['ZoneUpper_fibo_60']=( StockL_df['P_fibo_60']<= 20)
  StockL_df['ZoneUpper_fibo_100']=( StockL_df['P_fibo_100']<= 20)
  StockL_df['ZoneUpper_fibo_120']=( StockL_df['P_fibo_120']<= 20)
  StockL_df['ZoneUpper_fibo_200']=( StockL_df['P_fibo_200']<= 20)
  StockL_df['ZoneUpper_fibo_240']=( StockL_df['P_fibo_240']<= 20)
  StockL_df['ZoneUpper_fibo_600']=( StockL_df['P_fibo_600']<= 20)

  StockL_df['Zone236_fibo_03']=( StockL_df['P_fibo_03']>20)&( StockL_df['P_fibo_03']<26)
  StockL_df['Zone236_fibo_05']=( StockL_df['P_fibo_05']>20)&( StockL_df['P_fibo_05']<26)
  StockL_df['Zone236_fibo_10']=( StockL_df['P_fibo_10']>20)&( StockL_df['P_fibo_10']<26)
  StockL_df['Zone236_fibo_20']=( StockL_df['P_fibo_20']>20)&( StockL_df['P_fibo_20']<26)
  StockL_df['Zone236_fibo_25']=( StockL_df['P_fibo_25']>20)&( StockL_df['P_fibo_25']<26)
  StockL_df['Zone236_fibo_60']=( StockL_df['P_fibo_60']>20)&( StockL_df['P_fibo_60']<26)
  StockL_df['Zone236_fibo_100']=( StockL_df['P_fibo_100']>20)&( StockL_df['P_fibo_100']<26)
  StockL_df['Zone236_fibo_120']=( StockL_df['P_fibo_120']>20)&( StockL_df['P_fibo_120']<26)
  StockL_df['Zone236_fibo_200']=( StockL_df['P_fibo_200']>20)&( StockL_df['P_fibo_200']<26)
  StockL_df['Zone236_fibo_240']=( StockL_df['P_fibo_240']>20)&( StockL_df['P_fibo_240']<26)
  StockL_df['Zone236_fibo_600']=( StockL_df['P_fibo_600']>20)&( StockL_df['P_fibo_600']<26)

  StockL_df['Zone382_fibo_03']=( StockL_df['P_fibo_03']>35)&( StockL_df['P_fibo_03']<41)
  StockL_df['Zone382_fibo_05']=( StockL_df['P_fibo_05']>35)&( StockL_df['P_fibo_05']<41)
  StockL_df['Zone382_fibo_10']=( StockL_df['P_fibo_10']>35)&( StockL_df['P_fibo_10']<41)
  StockL_df['Zone382_fibo_20']=( StockL_df['P_fibo_20']>35)&( StockL_df['P_fibo_20']<41)
  StockL_df['Zone382_fibo_25']=( StockL_df['P_fibo_25']>35)&( StockL_df['P_fibo_25']<41)
  StockL_df['Zone382_fibo_60']=( StockL_df['P_fibo_60']>35)&( StockL_df['P_fibo_60']<41)
  StockL_df['Zone382_fibo_100']=( StockL_df['P_fibo_100']>35)&( StockL_df['P_fibo_100']<41)
  StockL_df['Zone382_fibo_120']=( StockL_df['P_fibo_120']>35)&( StockL_df['P_fibo_120']<41)
  StockL_df['Zone382_fibo_200']=( StockL_df['P_fibo_200']>35)&( StockL_df['P_fibo_200']<41)
  StockL_df['Zone382_fibo_240']=( StockL_df['P_fibo_240']>35)&( StockL_df['P_fibo_240']<41)
  StockL_df['Zone382_fibo_600']=( StockL_df['P_fibo_600']>35)&( StockL_df['P_fibo_600']<41)

  StockL_df['Zone50_fibo_03']=( StockL_df['P_fibo_03']>47)&( StockL_df['P_fibo_03']<53)
  StockL_df['Zone50_fibo_05']=( StockL_df['P_fibo_05']>47)&( StockL_df['P_fibo_05']<53)
  StockL_df['Zone50_fibo_10']=( StockL_df['P_fibo_10']>47)&( StockL_df['P_fibo_10']<53)
  StockL_df['Zone50_fibo_20']=( StockL_df['P_fibo_20']>47)&( StockL_df['P_fibo_20']<53)
  StockL_df['Zone50_fibo_25']=( StockL_df['P_fibo_25']>47)&( StockL_df['P_fibo_25']<53)
  StockL_df['Zone50_fibo_60']=( StockL_df['P_fibo_60']>47)&( StockL_df['P_fibo_60']<53)
  StockL_df['Zone50_fibo_100']=( StockL_df['P_fibo_100']>47)&( StockL_df['P_fibo_100']<53)
  StockL_df['Zone50_fibo_120']=( StockL_df['P_fibo_120']>47)&( StockL_df['P_fibo_120']<53)
  StockL_df['Zone50_fibo_200']=( StockL_df['P_fibo_200']>47)&( StockL_df['P_fibo_200']<53)
  StockL_df['Zone50_fibo_240']=( StockL_df['P_fibo_240']>47)&( StockL_df['P_fibo_240']<53)
  StockL_df['Zone50_fibo_600']=( StockL_df['P_fibo_600']>47)&( StockL_df['P_fibo_600']<53)

  StockL_df['Zone618_fibo_03']=( StockL_df['P_fibo_03']>58)&( StockL_df['P_fibo_03']<64)
  StockL_df['Zone618_fibo_05']=( StockL_df['P_fibo_05']>58)&( StockL_df['P_fibo_05']<64)
  StockL_df['Zone618_fibo_10']=( StockL_df['P_fibo_10']>58)&( StockL_df['P_fibo_10']<64)
  StockL_df['Zone618_fibo_20']=( StockL_df['P_fibo_20']>58)&( StockL_df['P_fibo_20']<64)
  StockL_df['Zone618_fibo_25']=( StockL_df['P_fibo_25']>58)&( StockL_df['P_fibo_25']<64)
  StockL_df['Zone618_fibo_60']=( StockL_df['P_fibo_60']>58)&( StockL_df['P_fibo_60']<64)
  StockL_df['Zone618_fibo_100']=( StockL_df['P_fibo_100']>58)&( StockL_df['P_fibo_100']<64)
  StockL_df['Zone618_fibo_120']=( StockL_df['P_fibo_120']>58)&( StockL_df['P_fibo_120']<64)
  StockL_df['Zone618_fibo_200']=( StockL_df['P_fibo_200']>58)&( StockL_df['P_fibo_200']<64)
  StockL_df['Zone618_fibo_240']=( StockL_df['P_fibo_240']>58)&( StockL_df['P_fibo_240']<64)
  StockL_df['Zone618_fibo_600']=( StockL_df['P_fibo_600']>58)&( StockL_df['P_fibo_600']<64)

  StockL_df['Zone786_fibo_03']=( StockL_df['P_fibo_03']>75)&( StockL_df['P_fibo_03']<81)
  StockL_df['Zone786_fibo_05']=( StockL_df['P_fibo_05']>75)&( StockL_df['P_fibo_05']<81)
  StockL_df['Zone786_fibo_10']=( StockL_df['P_fibo_10']>75)&( StockL_df['P_fibo_10']<81)
  StockL_df['Zone786_fibo_20']=( StockL_df['P_fibo_20']>75)&( StockL_df['P_fibo_20']<81)
  StockL_df['Zone786_fibo_25']=( StockL_df['P_fibo_25']>75)&( StockL_df['P_fibo_25']<81)
  StockL_df['Zone786_fibo_60']=( StockL_df['P_fibo_60']>75)&( StockL_df['P_fibo_60']<81)
  StockL_df['Zone786_fibo_100']=( StockL_df['P_fibo_100']>75)&( StockL_df['P_fibo_100']<81)
  StockL_df['Zone786_fibo_120']=( StockL_df['P_fibo_120']>75)&( StockL_df['P_fibo_120']<81)
  StockL_df['Zone786_fibo_200']=( StockL_df['P_fibo_200']>75)&( StockL_df['P_fibo_200']<81)
  StockL_df['Zone786_fibo_240']=( StockL_df['P_fibo_240']>75)&( StockL_df['P_fibo_240']<81)
  StockL_df['Zone786_fibo_600']=( StockL_df['P_fibo_600']>75)&( StockL_df['P_fibo_600']<81)

  StockL_df['ZoneLower_fibo_03']= (StockL_df['P_fibo_03']>=81)
  StockL_df['ZoneLower_fibo_05']= (StockL_df['P_fibo_05']>=81)
  StockL_df['ZoneLower_fibo_10']= (StockL_df['P_fibo_10']>=81)
  StockL_df['ZoneLower_fibo_20']= (StockL_df['P_fibo_20']>=81)
  StockL_df['ZoneLower_fibo_25']= (StockL_df['P_fibo_25']>=81)
  StockL_df['ZoneLower_fibo_60']= (StockL_df['P_fibo_60']>=81)
  StockL_df['ZoneLower_fibo_100']= (StockL_df['P_fibo_100']>=81)
  StockL_df['ZoneLower_fibo_120']= (StockL_df['P_fibo_120']>=81)
  StockL_df['ZoneLower_fibo_200']= (StockL_df['P_fibo_200']>=81)
  StockL_df['ZoneLower_fibo_240']= (StockL_df['P_fibo_240']>=81)
  StockL_df['ZoneLower_fibo_600']= (StockL_df['P_fibo_600']>=81)


  StockL_df['MA200Uptrend'] = (StockL_df['MA15'] > StockL_df['MA25']) & (StockL_df['MA25'] > StockL_df['MA60']) & (StockL_df['MA60']  > StockL_df['MA120'])   &(StockL_df['MA120']  > StockL_df['MA200'])
  StockL_df['MA200Downtrend'] =(StockL_df['MA15'] < StockL_df['MA25']) & (StockL_df['MA25'] < StockL_df['MA60'])& (StockL_df['MA60']  < StockL_df['MA120'])& (StockL_df['MA120']  < StockL_df['MA200'])
  StockL_df['CDC'] = (StockL_df['EMA12'] > StockL_df['EMA26'] )
  StockL_df['CDC_last'] = (StockL_df['EMA12-1'] < StockL_df['EMA26-1'])
  StockL_df['CDC_break'] = (StockL_df['EMA12']>StockL_df['EMA26'])&(StockL_df['EMA12-1']<StockL_df['EMA26-1'])
  StockL_df['CDC_break-1'] = StockL_df['CDC_break'].shift(1)
  StockL_df['CDC_break-2'] = StockL_df['CDC_break'].shift(2)

  StockL_df['EMA12_Turning'] = (StockL_df['EMA12%C'] > 0) & (StockL_df['EMA12%C-1'] < 0)
  # StockL_df['EMA26_Turning'] = (StockL_df['EMA26%Close'] > 0) & (StockL_df['EMA26%Close-1'] < 0)
  StockL_df['EMA12%'] = ((StockL_df['EMA12'] - StockL_df['EMA12-12']) / StockL_df['EMA12-12']) * 100
  StockL_df['EMA26%'] = ((StockL_df['EMA26'] - StockL_df['EMA26-26']) / StockL_df['EMA26-26']) * 100
  StockL_df['BB upper_band break'] = ((StockL_df['upper_band'] < StockL_df['close'])&(StockL_df['upper_band_1'] > StockL_df['close-1']))
  StockL_df['BB lower_band break'] = ((StockL_df['lower_band'] > StockL_df['close'])&(StockL_df['lower_band_1'] < StockL_df['close-1']))
  StockL_df['volmax25TF'] = (StockL_df['volume'] == StockL_df['VolMax25'])
  StockL_df['volmax60TF'] = (StockL_df['volume'] == StockL_df['VolMax60'])
  StockL_df['volmax100TF'] = (StockL_df['volume'] == StockL_df['VolMax100'])
  StockL_df['volmax200TF'] = (StockL_df['volume'] == StockL_df['VolMax200'])
  StockL_df['volmax600TF'] = (StockL_df['volume'] == StockL_df['VolMax600'])
  StockL_df['volmax25TF-1'] = (StockL_df['volume-1'] == StockL_df['VolMax-1_25'])
  StockL_df['volmax60TF-1'] = (StockL_df['volume-1'] == StockL_df['VolMax-1_60'])
  StockL_df['volmax100TF-1'] = (StockL_df['volume-1'] == StockL_df['VolMax-1_100'])
  StockL_df['volmax200TF-1'] = (StockL_df['volume-1'] == StockL_df['VolMax-1_200'])
  StockL_df['volmax600TF-1'] = (StockL_df['volume-1'] == StockL_df['VolMax-1_600'])
  StockL_df['volmax25TF-2'] = (StockL_df['volume-2'] == StockL_df['VolMax-2_25'])
  StockL_df['volmax60TF-2'] = (StockL_df['volume-2'] == StockL_df['VolMax-2_60'])
  StockL_df['volmax100TF-2'] = (StockL_df['volume-2'] == StockL_df['VolMax-2_100'])
  StockL_df['volmax200TF-2'] = (StockL_df['volume-2'] == StockL_df['VolMax-2_200'])
  StockL_df['volmax600TF-2'] = (StockL_df['volume-2'] == StockL_df['VolMax-2_600'])
  StockL_df['volmax25TF-3'] = (StockL_df['volume-3'] == StockL_df['VolMax-3_25'])
  StockL_df['volmax60TF-3'] = (StockL_df['volume-3'] == StockL_df['VolMax-3_60'])
  StockL_df['volmax100TF-3'] = (StockL_df['volume-3'] == StockL_df['VolMax-3_100'])
  StockL_df['volmax200TF-3'] = (StockL_df['volume-3'] == StockL_df['VolMax-3_200'])
  StockL_df['volmax600TF-3'] = (StockL_df['volume-3'] == StockL_df['VolMax-3_600'])
  StockL_df['volmax25TF-5'] = (StockL_df['volume-5'] == StockL_df['VolMax-5_25'])
  StockL_df['volmax60TF-5'] = (StockL_df['volume-5'] == StockL_df['VolMax-5_60'])
  StockL_df['volmax100TF-5'] = (StockL_df['volume-5'] == StockL_df['VolMax-5_100'])
  StockL_df['volmax200TF-5'] = (StockL_df['volume-5'] == StockL_df['VolMax-5_200'])
  StockL_df['volmax600TF-5'] = (StockL_df['volume-5'] == StockL_df['VolMax-5_600'])
  StockL_df['volmax25TF-10'] = (StockL_df['volume-10'] == StockL_df['VolMax-10_25'])
  StockL_df['volmax60TF-10'] = (StockL_df['volume-10'] == StockL_df['VolMax-10_60'])
  StockL_df['volmax100TF-10'] = (StockL_df['volume-10'] == StockL_df['VolMax-10_100'])
  StockL_df['volmax200TF-10'] = (StockL_df['volume-10'] == StockL_df['VolMax-10_200'])
  StockL_df['volmax600TF-10'] = (StockL_df['volume-10'] == StockL_df['VolMax-10_600'])

  StockL_df['Val25_60_TF'] = ((StockL_df['Values']>StockL_df['ValMA25'])&(StockL_df['Values']>StockL_df['ValMA60']))
  StockL_df['Vol25_60_TF'] = ((StockL_df['volume']>StockL_df['VolMA25'])&(StockL_df['volume']>StockL_df['VolMA60']))


# xxx

  StockL_df['%chgclose_ema12'] = ((StockL_df['close'] - StockL_df['EMA12'])/StockL_df['close']) *100
  StockL_df['%chgclose_ema26'] = ((StockL_df['close'] - StockL_df['EMA26'])/StockL_df['close']) *100

  StockL_df['EMA05_break'] = (StockL_df['close']>StockL_df['EMA05'])&(StockL_df['close-1']<StockL_df['EMA05-1'])
  StockL_df['EMA10_break'] = (StockL_df['close']>StockL_df['EMA10'])&(StockL_df['close-1']<StockL_df['EMA10-1'])
  StockL_df['EMA12_break'] = (StockL_df['close']>StockL_df['EMA12'])&(StockL_df['close-1']<StockL_df['EMA12-1'])
  StockL_df['EMA26_break'] = (StockL_df['close']>StockL_df['EMA26'])&(StockL_df['close-1']<StockL_df['EMA26-1'])
  StockL_df['EMA40_break'] = (StockL_df['close']>StockL_df['EMA40'])&(StockL_df['close-1']<StockL_df['EMA40-1'])
  StockL_df['EMA40_break_down'] = (StockL_df['close']<StockL_df['EMA40'])&(StockL_df['close-1']>StockL_df['EMA40-1'])
  StockL_df['EMA75_break'] = (StockL_df['close']>StockL_df['EMA75'])&(StockL_df['close-1']<StockL_df['EMA75-1'])
  StockL_df['EMA89_break'] = (StockL_df['close']>StockL_df['EMA89'])&(StockL_df['close-1']<StockL_df['EMA89-1'])
  StockL_df['EMA200_break'] = (StockL_df['close']>StockL_df['EMA200'])&(StockL_df['close-1']<StockL_df['EMA200-1'])

  StockL_df['EMA05_break-1'] = (StockL_df['close-1']>StockL_df['EMA05-1'])&(StockL_df['close-2']<StockL_df['EMA05-2'])
  StockL_df['EMA10_break-1'] = (StockL_df['close-1']>StockL_df['EMA10-1'])&(StockL_df['close-2']<StockL_df['EMA10-2'])
  StockL_df['EMA12_break-1'] = (StockL_df['close-1']>StockL_df['EMA12-1'])&(StockL_df['close-2']<StockL_df['EMA12-2'])
  StockL_df['EMA26_break-1'] = (StockL_df['close-1']>StockL_df['EMA26-1'])&(StockL_df['close-2']<StockL_df['EMA26-2'])
  StockL_df['EMA40_break-1'] = (StockL_df['close-1']>StockL_df['EMA40-1'])&(StockL_df['close-2']<StockL_df['EMA40-2'])
  StockL_df['EMA40_break_down-1'] = (StockL_df['close-1']<StockL_df['EMA40-1'])&(StockL_df['close-2']>StockL_df['EMA40-2'])
  StockL_df['EMA75_break-1'] = (StockL_df['close-1']>StockL_df['EMA75-1'])&(StockL_df['close-2']<StockL_df['EMA75-2'])
  StockL_df['EMA89_break-1'] = (StockL_df['close-1']>StockL_df['EMA89-1'])&(StockL_df['close-2']<StockL_df['EMA89-2'])
  StockL_df['EMA200_break-1'] = (StockL_df['close-1']>StockL_df['EMA200-1'])&(StockL_df['close-2']<StockL_df['EMA200-2'])

  StockL_df['EMA05_break-5'] = (StockL_df['close-5']>StockL_df['EMA05-5'])&(StockL_df['close-6']<StockL_df['EMA05-6'])
  StockL_df['EMA10_break-5'] = (StockL_df['close-5']>StockL_df['EMA10-5'])&(StockL_df['close-6']<StockL_df['EMA10-6'])
  StockL_df['EMA12_break-5'] = (StockL_df['close-5']>StockL_df['EMA12-5'])&(StockL_df['close-6']<StockL_df['EMA12-6'])
  StockL_df['EMA26_break-5'] = (StockL_df['close-5']>StockL_df['EMA26-5'])&(StockL_df['close-6']<StockL_df['EMA26-6'])
  StockL_df['EMA40_break-5'] = (StockL_df['close-5']>StockL_df['EMA40-5'])&(StockL_df['close-6']<StockL_df['EMA40-6'])
  StockL_df['EMA40_break_down-5'] = (StockL_df['close-5']<StockL_df['EMA40-5'])&(StockL_df['close-6']>StockL_df['EMA40-6'])
  StockL_df['EMA75_break-5'] = (StockL_df['close-5']>StockL_df['EMA75-5'])&(StockL_df['close-6']<StockL_df['EMA75-6'])
  StockL_df['EMA89_break-5'] = (StockL_df['close-5']>StockL_df['EMA89-5'])&(StockL_df['close-6']<StockL_df['EMA89-6'])
  StockL_df['EMA200_break-5'] = (StockL_df['close-5']>StockL_df['EMA200-5'])&(StockL_df['close-6']<StockL_df['EMA200-6'])


  StockL_df['EMA05_break-10'] = (StockL_df['close-10']>StockL_df['EMA05-10'])&(StockL_df['close-11']<StockL_df['EMA05-11'])
  StockL_df['EMA10_break-10'] = (StockL_df['close-10']>StockL_df['EMA10-10'])&(StockL_df['close-11']<StockL_df['EMA10-11'])
  StockL_df['EMA12_break-10'] = (StockL_df['close-10']>StockL_df['EMA12-10'])&(StockL_df['close-11']<StockL_df['EMA12-11'])
  StockL_df['EMA26_break-10'] = (StockL_df['close-10']>StockL_df['EMA26-10'])&(StockL_df['close-11']<StockL_df['EMA26-11'])
  StockL_df['EMA40_break-10'] = (StockL_df['close-10']>StockL_df['EMA40-10'])&(StockL_df['close-11']<StockL_df['EMA40-11'])
  StockL_df['EMA40_break_down-10'] = (StockL_df['close-10']>StockL_df['EMA40-10'])&(StockL_df['close-11']<StockL_df['EMA40-11'])
  StockL_df['EMA75_break-10'] = (StockL_df['close-10']>StockL_df['EMA75-10'])&(StockL_df['close-11']<StockL_df['EMA75-11'])
  StockL_df['EMA89_break-10'] = (StockL_df['close-10']>StockL_df['EMA89-10'])&(StockL_df['close-11']<StockL_df['EMA89-11'])
  StockL_df['EMA200_break-10'] = (StockL_df['close-10']>StockL_df['EMA200-10'])&(StockL_df['close-11']<StockL_df['EMA200-11'])

  # StockL_df['Gap%VP600'] = ((StockL_df['close'] - StockL_df['VP600']) / StockL_df['close']) * 100
  # StockL_df['Gap%VP200'] = ((StockL_df['close'] - StockL_df['VP200']) / StockL_df['close']) * 100
  # StockL_df['Gap%VP60'] = ((StockL_df['close'] - StockL_df['VP60']) / StockL_df['close']) * 100
  # StockL_df['Gap%VP25'] = ((StockL_df['close'] - StockL_df['VP25']) / StockL_df['close']) * 100
  # StockL_df['EMA12%-EMA26%'] =(StockL_df['EMA12%']- StockL_df['EMA26%'])

  StockL_df['CDC_break_down'] = (StockL_df['EMA12']<StockL_df['EMA26'])&(StockL_df['EMA12-1']>StockL_df['EMA26-1'])
  StockL_df['EMA12_break_down'] = (StockL_df['close']<StockL_df['EMA12'])&(StockL_df['close-1']>StockL_df['EMA12-1'])
  StockL_df['EMA26_break_down'] = (StockL_df['close']<StockL_df['EMA26'])&(StockL_df['close-1']>StockL_df['EMA26-1'])
  StockL_df['EMA75_break_down'] = (StockL_df['close']<StockL_df['EMA75'])&(StockL_df['close-1']>StockL_df['EMA75-1'])
  StockL_df['EMA89_break_down'] = (StockL_df['close']<StockL_df['EMA89'])&(StockL_df['close-1']>StockL_df['EMA89-1'])


  StockL_df['heikin_ashi_check'] = (StockL_df['EMA3']<StockL_df['EMA12'])&(StockL_df['close-1']>StockL_df['EMA12-1'])
  StockL_df['1st green EMA10 heikin_ashi_check'] = (StockL_df['HA_Close'] > StockL_df['HA_Open']) & (StockL_df['HA_Close-1'] < StockL_df['HA_Open-1']) & (StockL_df['HA_Close'] > StockL_df['HA_Open-1'] ) &  (StockL_df['PMin60%-1'] > 2 )  &  (StockL_df['HA_Close'] > StockL_df['EMA10'])
  StockL_df['1st green_heikin_ashi_check'] = (StockL_df['HA_Close'] > StockL_df['HA_Open']) & (StockL_df['HA_Close-1'] < StockL_df['HA_Open-1']) & (StockL_df['HA_Close'] > StockL_df['HA_Open-1'] )
  StockL_df['2nd green_heikin_ashi_check'] = (StockL_df['HA_Close'] > StockL_df['HA_Open']) & (StockL_df['HA_Close-1'] > StockL_df['HA_Open-1']) & (StockL_df['HA_Close-2'] < StockL_df['HA_Open-2']) & (StockL_df['HA_Close'] > StockL_df['HA_Open'])
  StockL_df['1st green_heikin_ashi_check-1'] = StockL_df['1st green EMA10 heikin_ashi_check'].shift(1)
  StockL_df['1st green_heikin_ashi_check-2'] = StockL_df['1st green EMA10 heikin_ashi_check'].shift(2)
  StockL_df['2nd green_heikin_ashi_check-1'] = StockL_df['2nd green_heikin_ashi_check'].shift(1)
  StockL_df['2nd green_heikin_ashi_check-2'] = StockL_df['2nd green_heikin_ashi_check'].shift(2)



  StockL_df.sort_values (by=['Mcap'], ascending=False, inplace=True)
  StockL_df['rank Mcap'] = range(1, len(StockL_df) + 1)

  StockL_df.sort_values (by=['Values'], ascending=False, inplace=True)
  StockL_df['rank Values'] = range(1, len(StockL_df) + 1)


  StockL_df.sort_values (by=['retune_chg'], ascending=False, inplace=True)
  StockL_df['rank retune_chg'] = range(1, len(StockL_df) + 1)
  return table2

def addltabelCDC(StockL_df):
  df_CDC =  StockL_df[(StockL_df['CDC'] == 'TRUE')]

  return df_CDC

def addltabelzoneRSI(StockL_df):
  df_CDC =  StockL_df[(StockL_df['zoneRSI'] == 'TRUE')]

  return df_CDC


########################### Graph table section ###################################
# add daily data for graph
def addtablegraphma2560bb_d1(stcknr_df):
  stcknr_df['datetime_'] = stcknr_df.index.date
  for i in ['open', 'high', 'close', 'low']: # loop stock for change type
    stcknr_df[i] = stcknr_df[i].astype('float64')
  # cal indictor
  stcknr_df['MA25'] = stcknr_df['close'].rolling(window=25).mean()
  stcknr_df['MA60'] = stcknr_df['close'].rolling(window=60).mean()
  stcknr_df['MA25vol'] = stcknr_df['volume'].rolling(window=20).mean()
  stcknr_df['MA60vol'] = stcknr_df['volume'].rolling(window=60).mean()
  stcknr_df['rstd'] = stcknr_df['close'].rolling(window=25).std()

  stcknr_df['upper_band'] = stcknr_df['MA25'] + 2 * stcknr_df['rstd']
  stcknr_df['lower_band'] = stcknr_df['MA25'] - 2 * stcknr_df['rstd']

  stcknr_df['close-1'] = stcknr_df['close'].shift(1)
  # stcknr_df['datetime_'] = stcknr_df.index.date
  stcknr_df['chg%'] = ((stcknr_df['close'] - stcknr_df['close-1'])/stcknr_df['close-1'])*100

  stcknr_df['ADX'] = ta.trend.adx(stcknr_df['high'],stcknr_df['low'],stcknr_df["close"],14,False)
  stcknr_df['DI+'] = ta.trend.adx_pos(stcknr_df['high'],stcknr_df['low'],stcknr_df["close"],14,False)
  stcknr_df['DI-'] = ta.trend.adx_neg(stcknr_df['high'],stcknr_df['low'],stcknr_df["close"],14,False)
  stcknr_df['ADX_Hline1'] = 30
  stcknr_df['ADX_Hline2'] = 45


# add week data for graph
def addtablegraphma2560bb_w(stcknrw_df):
  stcknrw_df['datetime_'] = stcknrw_df.index.date

  for i in ['open', 'high', 'close', 'low']: # loop stock for change type
    stcknrw_df[i] = stcknrw_df[i].astype('float64')
  # # cal indictor
  stcknrw_df['MA5'] = stcknrw_df['close'].rolling(window=5).mean()
  stcknrw_df['MA15'] = stcknrw_df['close'].rolling(window=15).mean()
  stcknrw_df['MA5vol'] = stcknrw_df['volume'].rolling(window=5).mean()
  stcknrw_df['MA15vol'] = stcknrw_df['volume'].rolling(window=15).mean()

  stcknrw_df['close-1'] = stcknrw_df['close'].shift(1)
  # stcknr_df['datetime_'] = stcknr_df.index.date
  stcknrw_df['chg%'] = ((stcknrw_df['close'] - stcknrw_df['close-1'])/stcknrw_df['close-1'])*100

  lastw_row = stcknrw_df.tail(1)
  lStockw_df = lastw_row.append(lastw_row,ignore_index=True)

  lv_syb = lv_stocknameresult
  lv_chg = lStockw_df._get_value(0,'chg%')
  lv_chg = str(lv_chg)
  lv_dt = str(lStockw_df._get_value(0,'datetime_'))



########################### Graph section ###################################
def addgraphma2560bb_d1(stcknr_df):
  pio.templates.default = "plotly_white"
  fig = make_subplots(rows=3, cols=1, shared_xaxes=True,vertical_spacing=0.01,row_heights=[0.8, 0.15,0.15]  ) # make sub plot use x aexs shear and use heights by ratio

  # add OHLC trace
  fig1 = fig.add_trace(go.Candlestick(x=stcknr_df.index,
                              open=stcknr_df['open'],
                              high=stcknr_df['high'],
                              low=stcknr_df['low'],
                              close=stcknr_df['close'],
                              showlegend=False), row=1, col=1)
  # add moving average traces
  fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['MA60'],
                          opacity=0.7,
                          line=dict(color='blue', width=2),
                          name='MA60'), row=1, col=1)

  fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['MA25'],
                          opacity=0.7,
                          line=dict(color='orange', width=2),
                          name='MA 25'), row=1, col=1)



  fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                         y=stcknr_df['lower_band'],
                         name='Lower Band',
                         line_color='rgba(173,204,255,1)'
                        ))
  fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                         y=stcknr_df['upper_band'],
                         name='Upper Band',
                         fill='tonexty',
                         fillcolor='rgba(173,204,255,0.2)',
                         line_color='rgba(173,204,255,1)'
                        ))

  fig1.update_layout(width=1600, height=800)
  # Plot volume trace on 2nd row
  fig3 = fig.add_trace(go.Bar(x=stcknr_df.index,
                      y=stcknr_df['volume']
                      ), row=3, col=1)
  # add moving average traces
  fig3 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['MA60vol'],
                          opacity=0.7,
                          line=dict(color='red', width=2),
                          name='MA60vol'), row=3, col=1)
  fig3 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['MA25vol'],
                          opacity=0.7,
                          line=dict(color='green', width=2),
                          name='MA25vol'), row=3, col=1)




  fig3.update_layout(width=1600, height=400)

  # Plot ADX trace on 2nd row
  fig2 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['ADX'],
                          line=dict(color='Black', width=2),
                          name='ADX'), row=2, col=1)

  fig2 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['DI+'],
                          opacity=.5,
                          line=dict(color='Green', width=2),
                          name='DI+'), row=2, col=1)
  fig2 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['DI-'],
                          opacity=.5,
                          line=dict(color='Red', width=2),
                          name='DI-'), row=2, col=1)

  fig2 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['ADX_Hline2'],
                          opacity=.5,
                          line=dict(color='gray', width=2),
                          name='45'), row=2, col=1)

  fig2 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                          y=stcknr_df['ADX_Hline1'],
                          opacity=.5,
                          line=dict(color='gray', width=2),
                          name='25'), row=2, col=1)


  fig2.update_layout(width=1600, height=400)

  # remove rangeslider
  fig.update_layout(xaxis_rangeslider_visible=False)
  # add chart title
  lv_titel = lv_stocknameresult +" Date:"+lv_dt+"<br>"+ lv_chg +"%"+"<br>"+lv_cond
  fig.update_layout(title = lv_titel)
  fig.update_layout(width=1600, height=1024)

  # build complete timepline from start date to end date
  dt_all = pd.date_range(start=stcknr_df.index[0],end=stcknr_df.index[-1], freq="5T")
  # retrieve the dates that ARE in the original datset
  dt_obs = [d.strftime("%Y-%m-%d") for d in pd.to_datetime(stcknr_df['datetime_'])]
  # define dates with missing values
  dt_breaks = [d for d in dt_all.strftime("%Y-%m-%d").tolist() if not d in dt_obs]
  # hide dates with no values
  fig.update_xaxes(rangebreaks=[dict(values=dt_breaks)])

  fig.show()


def addgraphma2560bb_w1(stcknrw_df):
  figw = make_subplots(rows=2, cols=1, shared_xaxes=True,vertical_spacing=0.01,row_heights=[0.8, 0.2]  ) # make sub plot use x aexs shear and use heights by ratio

  # add OHLC trace
  figw1 = figw.add_trace(go.Candlestick(x=stcknrw_df.index,
                              open=stcknrw_df['open'],
                              high=stcknrw_df['high'],
                              low=stcknrw_df['low'],
                              close=stcknrw_df['close'],
                              showlegend=False), row=1, col=1)
  # add moving average traces
  figw1 = figw.add_trace(go.Scatter(x=stcknrw_df.index,
                          y=stcknrw_df['MA15'],
                          opacity=0.7,
                          line=dict(color='blue', width=2),
                          name='MA15'), row=1, col=1)
  figw1 = figw.add_trace(go.Scatter(x=stcknrw_df.index,
                          y=stcknrw_df['MA5'],
                          opacity=0.7,
                          line=dict(color='orange', width=2),
                          name='MA 5'), row=1, col=1)
  figw1.update_layout(width=1600, height=800)
  # Plot volume trace on 2nd row
  figw2 = figw.add_trace(go.Bar(x=stcknrw_df.index,
                      y=stcknr_df['volume']
                      ), row=2, col=1)
  # add moving average traces
  figw2 = figw.add_trace(go.Scatter(x=stcknrw_df.index,
                          y=stcknr_df['MA60vol'],
                          opacity=0.7,
                          line=dict(color='red', width=2),
                          name='MA60vol'), row=2, col=1)
  figw2 = figw.add_trace(go.Scatter(x=stcknrw_df.index,
                          y=stcknr_df['MA25vol'],
                          opacity=0.7,
                          line=dict(color='green', width=2),
                          name='MA25vol'), row=2, col=1)
  figw2.update_layout(width=1600, height=400)
  # remove rangeslider
  figw.update_layout(xaxis_rangeslider_visible=False)
  # add chart title
  lv_titel = lv_stocknameresult +" Week:"+lv_dt+"<br>"+ lv_chg +"%"+"<br>"+lv_cond
  figw.update_layout(title = lv_titel)
  figw.update_layout(width=1600, height=1024)

  figw.show()

def addgraphmadonc_d1(stcknr_df):
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True,vertical_spacing=0.01,row_heights=[0.8, 0.2]  ) # make sub plot use x aexs shear and use heights by ratio

    # add OHLC trace
    fig1 = fig.add_trace(go.Candlestick(x=stcknr_df.index,
                                open=stcknr_df['open'],
                                high=stcknr_df['high'],
                                low=stcknr_df['low'],
                                close=stcknr_df['close'],
                                showlegend=False), row=1, col=1)
    # add moving average traces
    fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                            y=stcknr_df['uc'],
                            opacity=0.7,
                            line=dict(color='blue', width=2),
                            name='Hi'), row=1, col=1)

    fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                            y=stcknr_df['lc'],
                            opacity=0.7,
                            line=dict(color='blue', width=2),
                            name='Lo'), row=1, col=1)

    fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                            y=stcknr_df['mc'],
                            opacity=0.7,
                            line=dict(color='orange', width=2),
                            name='Mc'), row=1, col=1)

    fig1 = fig.add_trace(go.Scatter(x=stcknr_df.index,
                            y=stcknr_df['MA25'],
                            opacity=0.7,
                            line=dict(color='Black', width=2),
                            name='MA 25'), row=1, col=1)

    fig1.update_layout(width=1600, height=800)
    # Plot volume trace on 2nd row
    fig2 = fig.add_trace(go.Bar(x=stcknr_df['datetime_'],#x=stcknr_df.index,
                        y=stcknr_df['volume']
                        ), row=2, col=1)

    fig2.update_layout(width=1600, height=400)
    # remove rangeslider
    fig.update_layout(xaxis_rangeslider_visible=False)
    # add chart title
    lv_titel = lv_stocknameresult +" Date:"+lv_dt+"<br>"+ lv_chg +"%"+"<br>"+lv_cond
    fig.update_layout(title = lv_titel)
    fig.update_layout(width=1600, height=1024)

    # build complete timepline from start date to end date
    dt_all = pd.date_range(start=stcknr_df.index[0],end=stcknr_df.index[-1], freq="5T")
    # retrieve the dates that ARE in the original datset
    dt_obs = [d.strftime("%Y-%m-%d") for d in pd.to_datetime(stcknr_df['datetime_'])]
    # define dates with missing values
    dt_breaks = [d for d in dt_all.strftime("%Y-%m-%d").tolist() if not d in dt_obs]
    # hide dates with no values
    fig.update_xaxes(rangebreaks=[dict(values=dt_breaks)])

    fig.show()

def volume_profile(stcknr_df, return_raw=False):
    """
    create volume profile
    :param df: time-indexed HOLCV bar or time-indexed P-V tick
    :param price_pace: price bucket, default 5 cents
    :param return_raw: return raw data or figure
    :return: raw data or figure obj
    """
    cmin = min(stcknr_df.close)
    cmax = max(stcknr_df.close)
    if cmin < 2:
      price_pace = 0.01
    elif cmin >= 2 and cmin <5 :
        price_pace = 0.02
    elif cmin >= 5 and cmin <10 :
        price_pace = 0.05
    elif cmin >= 10 and cmin <25 :
        price_pace = 0.1
    elif cmin >= 25 and cmin <100 :
        price_pace = 0.25
    elif cmin >= 100 and cmin <200 :
        price_pace = 0.5
    elif cmin >= 200 and cmin <400 :
        price_pace = 1
    elif cmin >= 400  :
        price_pace = 2

    cmin_int = int(cmin / price_pace) * price_pace  # int(0.9) = 0
    cmax_int = int(cmax / price_pace) * price_pace
    if cmax_int < cmax:
        cmax_int += price_pace
    cmax_int += price_pace  # right bracket is not included in arrange

    price_buckets = np.arange(cmin_int, cmax_int, price_pace)
    price_coors = pd.Series(price_buckets).rolling(2).mean().dropna()
    vol_bars = np.histogram(stcknr_df.close, bins=price_buckets, weights=stcknr_df.volume)[0]

    if return_raw:
        return (price_coors.values, vol_bars)



    fig1 = go.Candlestick(
        x=stcknr_df.index,
        open=stcknr_df.open,
        high=stcknr_df.high,
        low=stcknr_df.low,
        close=stcknr_df.close,
        xaxis='x',
        yaxis='y2',
        visible=True,
        showlegend=False
    )

    fig2 = go.Bar(
        x=stcknr_df.index,
        y=stcknr_df.volume,
        yaxis='y',
        name='Volume',
        marker_color='red',
        showlegend=False
    )

    fig3 = go.Bar(
        x=vol_bars,
        y=price_coors.values,
        orientation='h',
        xaxis='x2',
        yaxis='y3',
        visible=True,
        showlegend=False,
        marker_color='blue',
        opacity=0.4
    )

    low = cmin_int
    high = cmax_int
    layout = go.Layout(
        title=go.layout.Title(text= Namestock + "_Volume Profile_" + date_formated),
        xaxis=go.layout.XAxis(
            side="bottom",
            title="Date",
            rangeslider=go.layout.xaxis.Rangeslider(visible=False)
        ),
        yaxis=go.layout.YAxis(
            side="right",
            title='Volume',
            showticklabels=False,
            domain=[0, 0.2]
        ),
        yaxis2=go.layout.YAxis(
            side="right",
            title='Price',
            range=[low, high],
            domain=[0.2, 1.0]
        ),
        xaxis2=go.layout.XAxis(
            side="top",
            showgrid=False,
            # volume bar on the right
            # unfortunately reversed is an auto-range
            # one solution is to add an invisible bar.
            # https://community.plotly.com/t/reversed-axis-with-range-specified/3806
            # autorange='reversed',
            ticks='',
            showticklabels=False,
            range=[0, int(vol_bars.max() * 5)],
            overlaying="x"
        ),
        yaxis3=go.layout.YAxis(
            side="left",
            range=[low, high],
            showticklabels=False,
            overlaying="y2"
        ),
    )




    fig = go.Figure(data=[fig1, fig2, fig3], layout=layout)

    # build complete timepline from start date to end date
    dt_all = pd.date_range(start=stcknr_df.index[0],end=stcknr_df.index[-1], freq="5T")
    # retrieve the dates that ARE in the original datset
    dt_obs = [d.strftime("%Y-%m-%d") for d in pd.to_datetime(stcknr_df['datetime_'])]
    # define dates with missing values
    dt_breaks = [d for d in dt_all.strftime("%Y-%m-%d").tolist() if not d in dt_obs]
    # hide dates with no values
    fig.update_xaxes(rangebreaks=[dict(values=dt_breaks)])
    fig.update_layout(width=1600, height=1024)
    fig.show()
    return fig


def addgrahehikenashi(Stock_df):

  # คำนวณ Heikin Ashi
  Stock_df['HA_Close'] = (Stock_df['Open'] + Stock_df['High'] + Stock_df['Low'] + Stock_df['Close']) / 4

  # สร้างคอลัมน์ HA_Open (ใช้ค่าเริ่มต้น)
  for i in range(Stock_df.shape[0]):
    if i > 0:
      Stock_df.loc[Stock_df.index[i],'HA_Open'] = (Stock_df['HA_Open'][i-1] + Stock_df['HA_Close'][i-1])/2
    elif i == 0:
      Stock_df.loc[Stock_df.index[0],'HA_Open'] = (Stock_df['Open'][0] + Stock_df['Close'][0])/2

  # คำนวณคอลัมน์ HA_High และ HA_Low
  Stock_df['HA_High'] = Stock_df[['High', 'HA_Open', 'HA_Close']].max(axis=1)
  Stock_df['HA_Low'] = Stock_df[['Low', 'HA_Open', 'HA_Close']].min(axis=1)
```
